## 1. What is OOP? (📘)

Imagine building with LEGO. Instead of a pile of individual bricks (procedural programming), OOP lets you create **pre-assembled units** (a car, a house, a robot), each with:
- **Data (attributes)**: information about the object, e.g. a car's colour, model, speed
- **Actions (methods)**: things it can do, e.g. accelerate, brake, turn

### Why bother?
- **Organisation**: large projects become manageable
- **Reusability**: one class → many objects
- **Easier debugging**: problems live inside self-contained units
- **Real-world modelling**: `Customer`, `Order`, `Invoice` map naturally to code

### You've been using objects all along (📗)

````python
lst = [1, 2, 3]
print(lst.count(2))             # a list object with a method
print(type(1), type([]), type(()), type({}), type(""))
````

In Python, **everything is an object**: numbers, strings, functions, even classes.

## 2. The four pillars (📘)

| Pillar | Idea | Car analogy |
|---|---|---|
| **Abstraction** | Show only what's essential; hide complexity | You use the pedals without knowing the engine internals |
| **Encapsulation** | Bundle data + methods; control access | The engine is inside a protective casing |
| **Inheritance** | Build new classes from existing ones | `SportsCar` inherits from `Car` and adds a spoiler |
| **Polymorphism** | Same method name, different behaviour | `Dog.speak()` barks, `Cat.speak()` meows |

## 4. The constructor `__init__` and `self`

`__init__` runs **automatically** when you create an object. Its job is to set up the object's initial **state** (its attributes).

`self` means "**this particular object**". It's always the first parameter of an instance method; Python passes it for you.

````python
class Dog:
    def __init__(self, breed):
        self.breed = breed          # an instance attribute

sam = Dog(breed="Lab")
frank = Dog(breed="Huskie")
print(sam.breed, frank.breed)
````

What happens on `Dog("Lab")`:
1. Python creates a new, empty `Dog` object
2. Calls `Dog.__init__(that_new_object, "Lab")`, so `self` *is* the new object
3. `self.breed = "Lab"` stores the attribute on that object
4. Returns the object, which gets assigned to `sam`

> `self` is a convention, not a keyword, but always use it.

## 5. Instance attributes vs class attributes

- **Instance attributes** (`self.name`): specific to each object; usually set in `__init__`
- **Class attributes**: shared by **all** instances; defined directly in the class body

````python
class Dog:
    species = "Canis familiaris"     # class attribute (same for all dogs)

    def __init__(self, name, breed):
        self.name = name             # instance attributes
        self.breed = breed

    def bark(self):                  # a method
        print(f"{self.name} says Woof!")

my_dog = Dog("Buddy", "Golden Retriever")
another_dog = Dog("Lucy", "Labrador")
print(my_dog.name, another_dog.breed)
my_dog.bark()
print(Dog.species, my_dog.species)   # accessible via the class or an instance
````

❗ Assigning through an instance creates an **instance** attribute that *shadows* the class attribute:

````python
another_dog.species = "Robot dog"
print(my_dog.species, another_dog.species, Dog.species)
print(vars(another_dog))            # the instance's own attributes
````

❌ **Pitfall:** a *mutable* class attribute is shared by everyone:

````python
class Team:
    members = []                    # ❌ shared list
    def add(self, name):
        self.members.append(name)

a, b = Team(), Team()
a.add("Asha")
print(b.members)                    # b sees Asha too!

class Team:
    def __init__(self):
        self.members = []           # ✅ each team gets its own list
````

## 6. Methods

Methods are functions defined inside a class that operate on the object's data.

````python
class Circle:
    pi = 3.14                                # class attribute

    def __init__(self, radius=1):
        self.radius = radius
        self.area = radius * radius * Circle.pi   # attributes can be computed

    def set_radius(self, new_radius):
        self.radius = new_radius
        self.area = new_radius * new_radius * self.pi

    def get_circumference(self):
        return self.radius * self.pi * 2

c = Circle()
print("Radius:", c.radius, "Area:", c.area, "Circumference:", c.get_circumference())
c.set_radius(2)
print("Radius:", c.radius, "Area:", c.area, "Circumference:", c.get_circumference())
````

`obj.method()` is just shorthand for `Class.method(obj)`:

````python
print(c.get_circumference(), Circle.get_circumference(c))
````

## 7. Default values in the constructor (📘)

````python
class Dog:
    def __init__(self, name="Unknown", breed="Mixed"):
        self.name = name
        self.breed = breed

dog1 = Dog()
dog2 = Dog("Rex")
dog3 = Dog("Bella", "Labrador")
for d in (dog1, dog2, dog3):
    print(d.name, d.breed)
````

## 8. Making objects print nicely: a preview of `__str__` / `__repr__`

````python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

b = Book("Python Rocks", "Jose", 200)
print(b)          # <__main__.Book object at 0x...>: not helpful

class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __repr__(self):
        return f"Book({self.title!r}, {self.author!r}, {self.pages})"

    def __str__(self):
        return f"{self.title} by {self.author}"

b = Book("Python Rocks", "Jose", 200)
print(b)          # uses __str__
print([b])        # containers use __repr__
````

## 9. A complete example: bank account

````python
class BankAccount:
    bank_name = "PyBank"                      # class attribute

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        self.transactions = []

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit must be positive")
            return
        self.balance += amount
        self.transactions.append(("deposit", amount))
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Funds Unavailable!")
            return
        self.balance -= amount
        self.transactions.append(("withdraw", amount))
        print(f"Withdrew {amount}. New balance: {self.balance}")

    def __str__(self):
        return f"Account owner: {self.owner}\nAccount balance: ${self.balance}"

acct = BankAccount("Jose", 100)
print(acct)
acct.deposit(50)
acct.withdraw(75)
acct.withdraw(500)
print(acct.transactions)
````

## 10. Introspection helpers

````python
print(hasattr(acct, "balance"), getattr(acct, "owner"), getattr(acct, "email", "n/a"))
setattr(acct, "email", "jose@pybank.com")
print(acct.email)
print([a for a in dir(acct) if not a.startswith("_")])
````

## 📝 Summary
- A **class** is a blueprint; an **object** is an instance of it.
- `__init__(self, ...)` initialises instance attributes; `self` is the current object.
- Class attributes are shared; instance attributes are per object.
- Avoid mutable class attributes for per-object data.
- Methods are functions that receive `self`.

import math

class Line:
    def __init__(self, coor1, coor2):
        self.coor1, self.coor2 = coor1, coor2
    def distance(self):
        (x1, y1), (x2, y2) = self.coor1, self.coor2
        return math.hypot(x2 - x1, y2 - y1)
    def slope(self):
        (x1, y1), (x2, y2) = self.coor1, self.coor2
        return (y2 - y1) / (x2 - x1)

class Cylinder:
    pi = math.pi
    def __init__(self, height=1, radius=1):
        self.height, self.radius = height, radius
    def volume(self):
        return self.pi * self.radius ** 2 * self.height
    def surface_area(self):
        return 2 * self.pi * self.radius * (self.radius + self.height)

class Student:
    def __init__(self, name, grades=None):
        self.name = name
        self.grades = grades if grades is not None else []
    def average(self):
        return sum(self.grades) / len(self.grades) if self.grades else 0
