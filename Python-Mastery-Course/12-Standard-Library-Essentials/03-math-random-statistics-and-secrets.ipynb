## Part 1: `math` (📗)

````python
import math

value = 4.35
print(math.floor(value), math.ceil(value), round(value))
print(round(4.5), round(5.5))           # banker's rounding: halves go to the even number
````

````python
print(math.pi, math.e, math.tau)
print(math.inf, -math.inf, math.nan)
````

### Logarithms

````python
print(math.e)
print(math.log(math.e))      # natural log (base e)
print(math.log(100, 10))     # custom base
print(math.log10(100))
print(math.log2(1024))
print(math.exp(2))           # e ** 2
````

### Trigonometry (angles in **radians**)

````python
print(math.sin(10), math.cos(10), math.tan(10))
print(math.degrees(math.pi / 2), math.radians(180))
print(round(math.sin(math.radians(30)), 10))
````

### Other useful functions

````python
print(math.sqrt(2), math.isqrt(17), math.cbrt(27))        # cbrt: 3.11+
print(math.factorial(6), math.comb(10, 3), math.perm(10, 3))
print(math.gcd(48, 18), math.lcm(4, 6))
print(math.hypot(3, 4), math.dist((1, 1), (4, 5)))
print(math.prod([2, 3, 4]), math.fsum([0.1] * 10), sum([0.1] * 10))   # fsum avoids float drift
print(math.isclose(0.1 + 0.2, 0.3), math.copysign(3, -1), math.trunc(-2.7))
````

## Part 2: `random` (📗)

`random` produces **pseudo-random** numbers: they look random but come from a deterministic algorithm.

### Seeds: reproducible randomness

````python
import random

random.seed(101)                    # the same seed → the same sequence every time
print(random.randint(0, 100), random.randint(0, 100), random.randint(0, 100))
random.seed(101)
print(random.randint(0, 100), random.randint(0, 100), random.randint(0, 100))
````

### Random integers & floats

````python
print(random.randint(1, 6))           # both ends INCLUSIVE
print(random.randrange(0, 100, 5))    # like range(): stop excluded
print(random.random())                # float in [0.0, 1.0)
print(random.uniform(1.5, 2.5))       # float in a range
````

### Random with sequences

````python
mylist = list(range(0, 20))
print(random.choice(mylist))                    # grab one random item
print(random.choices(mylist, k=10))             # sample WITH replacement (repeats possible)
print(random.sample(mylist, k=5))               # sample WITHOUT replacement (no repeats)
random.shuffle(mylist)                          # shuffle IN PLACE (returns None)
print(mylist)
````

````python
# Weighted choices
outcomes = ["common", "rare", "legendary"]
print(random.choices(outcomes, weights=[80, 18, 2], k=10))
````

### Random distributions (📗)

````python
print(random.uniform(a=0, b=100))                 # uniform distribution
print(random.gauss(mu=0, sigma=1))                # normal (Gaussian) distribution
samples = [random.gauss(170, 10) for _ in range(10_000)]
print(round(sum(samples) / len(samples), 1))      # ≈ 170
````

### Example: simulate dice rolls

````python
from collections import Counter
random.seed(7)
rolls = Counter(random.randint(1, 6) + random.randint(1, 6) for _ in range(10_000))
for total in range(2, 13):
    print(f"{total:>2} {'█' * (rolls[total] // 50)}")
````

## Part 3: `statistics`

````python
import statistics as stats

marks = [72, 85, 90, 66, 85, 78, 95, 85, 60]
print("mean   :", stats.mean(marks))
print("median :", stats.median(marks))
print("mode   :", stats.mode(marks))
print("stdev  :", round(stats.stdev(marks), 2))       # sample standard deviation
print("pstdev :", round(stats.pstdev(marks), 2))      # population standard deviation
print("variance:", round(stats.variance(marks), 2))
print("quartiles:", stats.quantiles(marks, n=4))
print("multimode:", stats.multimode([1, 1, 2, 2, 3]))

hours = [1, 2, 3, 4, 5]
score = [52, 60, 71, 79, 88]
print("correlation:", round(stats.correlation(hours, score), 3))
slope, intercept = stats.linear_regression(hours, score)
print(f"score ≈ {slope:.1f} × hours + {intercept:.1f}")
````

## Part 4: `secrets`: randomness for security

> ⚠️ **Never use `random` for passwords, tokens, OTPs or anything security-related.** It's predictable: if someone learns the internal state (or the seed), they can reproduce all "random" values. Use **`secrets`**, which draws from the operating system's cryptographically secure generator.

````python
import secrets, string

print(secrets.token_hex(16))                 # e.g. API keys
print(secrets.token_urlsafe(16))             # e.g. password-reset links
print(secrets.randbelow(1_000_000))          # a random int below the limit
otp = "".join(secrets.choice(string.digits) for _ in range(6))
print("OTP:", otp)

alphabet = string.ascii_letters + string.digits + string.punctuation
password = "".join(secrets.choice(alphabet) for _ in range(16))
print("Password:", password)

print(secrets.compare_digest("token123", "token123"))   # constant-time comparison (prevents timing attacks)
````
