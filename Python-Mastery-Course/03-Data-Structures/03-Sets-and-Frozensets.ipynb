## 1. Creating sets

````python
fruits = {"apple", "banana", "cherry"}
print(fruits)
nums = set([1, 1, 2, 2, 3, 3])     # duplicates are removed automatically
print(nums)
letters = set("mississippi")
print(letters)
empty_set = set()                  # ✅ the only way to make an empty set
empty_dict = {}                    # ❌ this is an empty DICT!
print(type(empty_set), type(empty_dict))
````

- Sets are **unordered**: there's no indexing (`fruits[0]` is an error), and the print order may vary.
- Items must be **hashable** (immutable): numbers, strings, tuples are OK; lists and dicts are not.

````python
{[1, 2], [3]}     # ❌ TypeError: unhashable type: 'list'
````

## 2. Adding and removing

````python
s = {1, 2, 3}
s.add(4)                 # add one item
s.add(2)                 # already there → nothing happens
print(s)
s.update([5, 6], {7})    # add many items from iterables
print(s)
s.remove(7)              # remove; KeyError if missing
s.discard(100)           # remove if present; NO error if missing
print(s)
item = s.pop()           # remove & return an arbitrary item
print(item, s)
c = s.copy()
s.clear()
print(s, c)
````

````python
{1, 2}.remove(10)        # ❌ KeyError: 10
````

## 3. Set operations

```
     A = {1, 2, 3}        B = {3, 4, 5}
   ┌─────────┬───┬─────────┐
   │  1  2   │ 3 │   4  5  │
   └─────────┴───┴─────────┘
     A - B    A & B   B - A
   A | B = everything      A ^ B = everything except the middle
```

| Operation | Operator | Method | Result for A, B |
|---|---|---|---|
| Union | `A \| B` | `A.union(B)` | `{1, 2, 3, 4, 5}` |
| Intersection | `A & B` | `A.intersection(B)` | `{3}` |
| Difference | `A - B` | `A.difference(B)` | `{1, 2}` |
| Symmetric difference | `A ^ B` | `A.symmetric_difference(B)` | `{1, 2, 4, 5}` |

````python
a = {1, 2, 3}
b = {3, 4, 5}
print(a | b, a.union(b))
print(a & b, a.intersection(b))
print(a - b, a.difference(b))
print(b - a)
print(a ^ b, a.symmetric_difference(b))
````

💡 The **methods** accept any iterable; the **operators** need both sides to be sets:

````python
print(a.union([7, 8], (9,)))   # works
try:
    a | [7, 8]
except TypeError as e:
    print("TypeError:", e)
````

### In-place versions (`_update` / augmented operators)

| Method | Operator | Effect on `s` |
|---|---|---|
| `s.update(t)` | `s \|= t` | add items of `t` |
| `s.intersection_update(t)` | `s &= t` | keep only items also in `t` |
| `s.difference_update(t)` | `s -= t` | remove items found in `t` |
| `s.symmetric_difference_update(t)` | `s ^= t` | keep items in exactly one |

````python
s1 = {1, 2, 3}; s1.difference_update({1, 4}); print(s1)
s2 = {1, 2, 3}; s2.intersection_update({1, 2, 4}); print(s2)
s3 = {1, 2, 3}; s3.symmetric_difference_update({1, 4}); print(s3)
s4 = {1, 2}; s4 |= {9}; print(s4)
````

## 4. Relationship tests

````python
small = {1, 2}
big = {1, 2, 3, 4}
other = {7, 8}
print(small.issubset(big), small <= big)       # every item of small is in big
print(small < big)                             # proper subset (not equal)
print(big.issuperset(small), big >= small)
print(small.isdisjoint(other))                 # no items in common
print({1, 2} == {2, 1})                        # order doesn't matter
````

## 5. Practical examples

````python
# 1) Remove duplicates
emails = ["a@x.com", "b@x.com", "a@x.com"]
print(set(emails), len(set(emails)))

# 2) Find common / unique items
python_devs = {"Asha", "Ravi", "Meena"}
java_devs = {"Ravi", "John"}
print("Both:", python_devs & java_devs)
print("Only Python:", python_devs - java_devs)
print("Exactly one:", python_devs ^ java_devs)

# 3) Unique words in a sentence
text = "the cat and the hat and the bat"
print(sorted(set(text.split())))

# 4) Set comprehension
print({n % 3 for n in range(10)})
````

## 6. Speed: why sets are great for membership tests

Sets use **hashing**, so `x in my_set` takes about the same time no matter how big the set is (O(1) on average). For a list, Python must scan every element (O(n)).

````python
import timeit
big_list = list(range(100_000))
big_set = set(big_list)
print("list:", timeit.timeit(lambda: 99_999 in big_list, number=200))
print("set: ", timeit.timeit(lambda: 99_999 in big_set, number=200))
````

## 7. `frozenset`: an immutable set

A `frozenset` can't be changed, so it is **hashable**: it can be a dict key or an item inside another set.

````python
fs = frozenset([1, 2, 3])
print(fs, fs | {4})              # operations return new frozensets
groups = {frozenset({"a", "b"}): "team 1"}
print(groups[frozenset({"b", "a"})])
````

````python
fs.add(4)    # ❌ AttributeError: 'frozenset' object has no attribute 'add'
````

## 📝 Summary
