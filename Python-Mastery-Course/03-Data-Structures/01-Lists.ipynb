## 1. What is a list?

A list is an **ordered, mutable** collection that can hold items of **any type**, including duplicates. Written with square brackets.

````python
numbers = [1, 2, 3, 4, 5]
mixed = [10, "hello", 3.14, True, None, [1, 2]]
empty = []
also_empty = list()
from_string = list("abc")
from_range = list(range(5))
print(numbers, mixed, empty, also_empty, from_string, from_range, sep="\n")
print(len(mixed))
````

## 2. Indexing & slicing (same rules as strings)

````python
my_list = ["one", "two", "three", 4, 5]
print(my_list[0])       # 'one'
print(my_list[-1])      # 5
print(my_list[1:])      # ['two', 'three', 4, 5]
print(my_list[:3])      # ['one', 'two', 'three']
print(my_list[::-1])    # reversed copy
print(my_list[::2])     # every second item
````

## 3. Lists are mutable

Unlike strings, you can change lists in place.

````python
colors = ["red", "green", "blue"]
colors[0] = "crimson"          # replace one item
print(colors)
colors[1:3] = ["lime", "navy", "teal"]   # replace a slice (sizes can differ)
print(colors)
del colors[0]                  # delete by index
print(colors)
del colors[:]                  # delete everything (same as clear())
print(colors)
````

### Concatenation and repetition create **new** lists

````python
my_list = ["one", "two"]
print(my_list + ["new item"])
print(my_list)                       # unchanged
my_list = my_list + ["add new item permanently"]
print(my_list)
print(my_list * 2)
````

## 4. List methods

### Adding items: `append`, `extend`, `insert`

````python
lst = [1, 2, 3]
lst.append(4)              # add ONE item at the end
print(lst)
lst.append([5, 6])         # ❗ appends the list as a single item
print(lst)

lst = [1, 2, 3]
lst.extend([4, 5])         # add EACH item from an iterable
print(lst)
lst.insert(1, 99)          # insert at index 1
print(lst)
lst.insert(100, "end")     # index past the end → appended
print(lst)
````

### Removing items: `remove`, `pop`, `clear`

````python
lst = ["a", "b", "c", "b"]
lst.remove("b")            # removes the FIRST matching VALUE
print(lst)
last = lst.pop()           # removes & returns the last item
print(last, lst)
first = lst.pop(0)         # removes & returns the item at index 0
print(first, lst)
lst.clear()
print(lst)
````

````python
[1, 2, 3].remove(10)       # ❌ ValueError: list.remove(x): x not in list
````

````python
[].pop()                   # ❌ IndexError: pop from empty list
````

### Searching: `index`, `count`, `in`

````python
lst = [1, 2, 3, 2, 2]
print(lst.index(2))         # first position of 2
print(lst.index(2, 2))      # search starting at index 2
print(lst.count(2))         # how many 2s
print(3 in lst)
````

### Ordering: `sort`, `reverse`

````python
new_list = ["a", "e", "x", "b", "c"]
new_list.sort()                    # sorts IN PLACE
print(new_list)
new_list.sort(reverse=True)
print(new_list)
nums = [3, 1, 2]
nums.reverse()                     # reverses IN PLACE (no sorting)
print(nums)
````

### ❌ Be careful with assignment! (📗 Advanced Lists)

In-place methods return `None`:

````python
x = [3, 1, 2]
result = x.sort()
print(result)     # None!
print(x)          # sorted
````

✅ Use the built-in **`sorted()`**, which returns a **new** list and works on any iterable:

````python
x = [3, 1, 2]
y = sorted(x)
print(x, y)
print(sorted("python"))
print(sorted({3, 1, 2}, reverse=True))
````

### Sorting with `key=`

`key` is a function applied to each item to decide the order.

````python
words = ["banana", "Apple", "cherry", "date"]
print(sorted(words))                      # uppercase first (code-point order)
print(sorted(words, key=str.lower))       # case-insensitive
print(sorted(words, key=len))             # by length
students = [("Asha", 88), ("Ravi", 95), ("Meena", 72)]
print(sorted(students, key=lambda s: s[1], reverse=True))   # by score
# sort by multiple criteria: length, then alphabetically
print(sorted(words, key=lambda w: (len(w), w.lower())))
````

Sorting is **stable**: items with equal keys keep their original order.

### Copying: `copy()`

````python
original = [1, 2, 3]
alias = original          # NOT a copy, same object
clone = original.copy()   # also: original[:] or list(original)
original.append(4)
print(alias, clone)
````

## 5. Useful built-ins with lists

````python
nums = [4, 8, 15, 16, 23, 42]
print(len(nums), sum(nums), min(nums), max(nums))
print(sum(nums) / len(nums))           # average
print(list(reversed(nums)))
print(any(n > 40 for n in nums), all(n > 0 for n in nums))
for i, n in enumerate(nums[:3]):
    print(i, n)
````

## 6. Nested lists & matrices

````python
lst_1 = [1, 2, 3]
lst_2 = [4, 5, 6]
lst_3 = [7, 8, 9]
matrix = [lst_1, lst_2, lst_3]
print(matrix)
print(matrix[0])        # first row
print(matrix[1][2])     # row 1, column 2 → 6
for row in matrix:
    print(" ".join(str(v) for v in row))
first_col = [row[0] for row in matrix]      # list comprehension (lesson 05)
print(first_col)
transposed = [list(col) for col in zip(*matrix)]
print(transposed)
````

❌ **Pitfall when creating a 2D list:**

````python
grid = [[0] * 3] * 3       # ❌ three references to the SAME inner list
grid[0][0] = 1
print(grid)
grid = [[0] * 3 for _ in range(3)]   # ✅ three independent lists
grid[0][0] = 1
print(grid)
````

## 7. Lists as stacks and queues

**Stack (LIFO):** `append()` + `pop()`, both fast.

````python
stack = []
stack.append("page1"); stack.append("page2"); stack.append("page3")
print(stack.pop(), stack)    # 'back' button
````

**Queue (FIFO):** `list.pop(0)` is slow for big lists (it shifts every item). Use `collections.deque`:

````python
from collections import deque
queue = deque(["Asha", "Ravi"])
queue.append("Meena")       # join at the back
print(queue.popleft())      # serve from the front → 'Asha'
print(queue)
````

## 8. Unpacking lists

````python
point = [3, 4]
x, y = point
head, *tail = [1, 2, 3, 4]
print(x, y, head, tail)
print([*point, 5, *tail])      # unpack into a new list
````

## 📝 Summary

| Method | Does | Returns |
|---|---|---|
| `append(x)` | add item at end | `None` |
| `extend(iter)` | add all items | `None` |
| `insert(i, x)` | insert at index | `None` |
| `remove(x)` | delete first `x` | `None` |
| `pop([i])` | delete at index (default last) | the item |
| `clear()` | delete all | `None` |
| `index(x)` | position of first `x` | `int` |
| `count(x)` | occurrences | `int` |
| `sort(key, reverse)` | sort in place | `None` |
| `reverse()` | reverse in place | `None` |
| `copy()` | shallow copy | new list |
