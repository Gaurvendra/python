## 1. From loop to comprehension

General form: `[expression for item in iterable]`

````python
# Classic loop
squares = []
for x in range(10):
    squares.append(x ** 2)
print(squares)

# Same thing as a list comprehension
squares = [x ** 2 for x in range(10)]
print(squares)
````

### 📗 Examples from the bootcamp

````python
# Example 1: every letter in a string
lst = [letter for letter in "word"]
print(lst)

# Example 2: square numbers
print([x ** 2 for x in range(0, 11)])

# Example 3: only even numbers (filter with if)
print([x for x in range(11) if x % 2 == 0])

# Example 4: Celsius → Fahrenheit
celsius = [0, 10, 20.1, 34.5]
fahrenheit = [(9 / 5) * temp + 32 for temp in celsius]
print(fahrenheit)

# Example 5: nested comprehension
print([x ** 2 for x in [x ** 2 for x in range(11)]])
````

## 2. Filtering: `if` at the END

`[expression for item in iterable if condition]`, which keeps only matching items.

````python
words = ["apple", "kiwi", "banana", "fig", "cherry"]
print([w for w in words if len(w) > 4])
print([w.upper() for w in words if "a" in w])
````

## 3. Transforming: `if ... else` at the FRONT

`[a if condition else b for item in iterable]`, which decides what to output for **every** item.

````python
nums = range(1, 11)
print(["even" if n % 2 == 0 else "odd" for n in nums])
print([n if n > 0 else 0 for n in [-2, 5, -1, 3]])     # clip negatives
````

> ❌ **Common mistake:** `[x for x in nums if x > 0 else 0]` is a SyntaxError. A filter `if` (no `else`) goes at the **end**; a ternary `if/else` goes at the **front**.

## 4. Multiple `for` clauses (nested loops)

The `for`s are read **left to right, outer to inner**, same order as nested loops.

````python
pairs = [(x, y) for x in range(3) for y in range(2)]
print(pairs)
# equivalent to:
pairs = []
for x in range(3):
    for y in range(2):
        pairs.append((x, y))
print(pairs)
````

````python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [n for row in matrix for n in row]          # flatten
print(flat)
transposed = [[row[i] for row in matrix] for i in range(3)]
print(transposed)
table = [[i * j for j in range(1, 6)] for i in range(1, 6)]   # multiplication table
for row in table:
    print(row)
````

## 5. 🆕 Unpacking in comprehensions (Python 3.15, PEP 798)

Python 3.15 allows `*` and `**` directly in comprehensions to **flatten** iterables or **merge** dicts:

````python
lists = [[1, 2], [3, 4], [5]]
print([*L for L in lists])                # same as [x for L in lists for x in L]

dicts = [{"a": 1}, {"b": 2}, {"a": 3}]
print({**d for d in dicts})               # merge dicts, later keys win

sets = [{1, 2}, {2, 3}]
print({*s for s in sets})                 # union of sets
````

> On Python ≤ 3.14 these lines are a `SyntaxError`; use the double-`for` form instead.

## 6. Dictionary comprehensions

`{key_expr: value_expr for item in iterable}`

````python
print({x: x ** 2 for x in range(6)})
names = ["Asha", "Ravi", "Meena"]
print({name: len(name) for name in names})
print({i: name for i, name in enumerate(names, start=1)})
prices = {"apple": 100, "banana": 40, "mango": 150}
print({k: v * 0.9 for k, v in prices.items() if v > 50})   # 10% discount on items > 50
````

## 7. Set comprehensions

````python
words = ["Apple", "apple", "APPLE", "Banana"]
print({w.lower() for w in words})
print({len(w) for w in words})
````

## 8. Generator expressions: lazy, memory-efficient

Use **parentheses** instead of brackets. Items are produced **one at a time**, on demand, so nothing is stored in a big list.

````python
gen = (x ** 2 for x in range(5))
print(gen)
print(next(gen), next(gen))
print(list(gen))      # the remaining items
````

````python
# Perfect for feeding into functions like sum/max/any/all (extra parentheses not needed):
print(sum(x ** 2 for x in range(1_000_000)))
print(any(n > 5 for n in [1, 3, 7]))

import sys
print(sys.getsizeof([x for x in range(100_000)]), "bytes for the list")
print(sys.getsizeof((x for x in range(100_000))), "bytes for the generator")
````

> ❗ There is **no "tuple comprehension"**: `(x for x in ...)` is a generator. Use `tuple(x for x in ...)` to build a tuple.

## 9. Comprehensions vs `map`/`filter`

````python
nums = [1, 2, 3, 4, 5]
print(list(map(lambda x: x * 2, nums)), [x * 2 for x in nums])
print(list(filter(lambda x: x % 2, nums)), [x for x in nums if x % 2])
````

Comprehensions are usually considered more readable (and are often slightly faster). `map` is fine when you already have a named function: `list(map(str, nums))`.

## 10. When NOT to use a comprehension
- When it gets too long or deeply nested: use a normal loop
- When you only want **side effects** (printing, appending elsewhere): use a loop
- ❌ `[print(x) for x in items]` builds a useless list of `None`s

## 📝 Summary

| Kind | Syntax |
|---|---|
| list | `[expr for x in it if cond]` |
| dict | `{k: v for x in it}` |
| set | `{expr for x in it}` |
| generator | `(expr for x in it)` |
| filter | `if` at the end |
| transform | `a if c else b` at the front |
| 🆕 3.15 flatten | `[*x for x in it]`, `{**d for d in it}` |
