## 🆕 Project 5: To-do CLI with tests

Build `todo`, a command-line to-do manager.

**Requirements**
- Commands: `todo add "Buy milk" --due 2026-11-01 --priority high`, `todo list [--all]`, `todo done 3`, `todo remove 3`
- Store tasks in a JSON file in the user's home folder (`Path.home() / ".todo.json"`)
- Model each task as a `@dataclass` with an `Enum` priority
- Print a neat table with f-strings (overdue tasks marked ⚠️)
- **pytest** tests for all core functions (use the `tmp_path` fixture for files)
- Package it with `pyproject.toml` and a `[project.scripts]` entry so `todo` works anywhere

**Stretch:** colours with `rich`; `todo stats` showing completion rate; type-check with mypy; lint with Ruff.

## 🆕 Project 6: Async web-page monitor

Watch a list of URLs and alert you when one changes or goes down.

**Requirements**
- Config in a TOML file: URLs, check interval, CSS selector to watch (optional)
- Fetch all URLs **concurrently** with `asyncio` + `httpx.AsyncClient` (limit concurrency with a `Semaphore`)
- Detect changes by hashing the content (`hashlib.sha256`)
- Log every check with `logging`; on change/failure send an email (`EmailMessage`) or print an alert
- Handle timeouts and errors per URL (one failure must not stop the others: `TaskGroup` + `except*`)

**Stretch:** store history in SQLite; a `--once` CLI flag; run it on a schedule.

## 🆕 Project 7: Expense tracker

**Requirements**
- SQLite database (`sqlite3`) with tables for expenses and categories
- Add, list and filter expenses by month/category (parameterised queries!)
- Monthly report: totals per category, mean/median spend (`statistics`), biggest expense
- Export the report to CSV and to a formatted Excel file (`openpyxl`)
- Money stored with `Decimal`, never floats

**Stretch:** import bank-statement CSVs; a chart with matplotlib; a simple web UI with Flask/FastAPI.

## 🆕 Project 8: Photo organiser

**Requirements**
- Scan a folder recursively (`Path.rglob`) for images
- Read the date taken from EXIF (`Image.getexif()`), falling back to the file's modified time
- Move/copy photos into `YYYY/MM/` folders (🆕 `Path.copy()` / `Path.move()` in 3.14)
- Create thumbnails and detect duplicates by content hash
- Dry-run mode that only prints what it *would* do (`argparse` flag)

**Stretch:** process images in parallel with `ProcessPoolExecutor`; a progress bar.

## 🆕 Project 9: Publish your own library

**Requirements**
- Create `yourname-textstats` with `uv init --lib` (or `poetry new`): functions for word counts, readability scores, top-N words
- Full type hints (`py.typed`), mypy/ty clean; docstrings; README with examples
- pytest + Hypothesis tests, coverage ≥ 90%; Ruff
- `pyproject.toml` with classifiers, URLs, `license = "MIT"`, an optional extra `[cli]` adding a `textstats` command
- Build the wheel & sdist, inspect them, `twine check`, publish to **TestPyPI**, install it in a fresh venv
- GitHub Actions: CI on 3.12–3.15 and a release workflow with **trusted publishing**

## 🆕 Project 10: Validated REST API

**Requirements**
- A "library management" API with FastAPI: books, members, loans
- Pydantic `Create`/`Update`/`Public` schemas with constraints, discriminated unions for loan events, custom types (ISBN)
- SQLAlchemy 2.x models, relationships, pagination, filtering; Alembic migrations
- `pydantic-settings` configuration from `.env`; structured JSON logging
- Tests with `TestClient` and dependency overrides; OpenAPI docs reviewed at `/docs`
- Dockerfile using `uv sync --locked`; run with uvicorn

## Tips for every project
1. **Plan first:** write the steps in plain English (or as comments) before coding.
2. **Start small:** get a minimal version working, then add features.
3. Use **Git** from day one (Module 16): commit after each working step.
4. Write **tests** for the logic (Module 07 · 03).
5. Put the code in `.py` files and a proper project folder, not only notebooks.
6. Share it on GitHub with a `README.md` explaining how to run it.
