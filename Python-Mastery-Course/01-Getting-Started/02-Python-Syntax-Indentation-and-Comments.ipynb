## 1. Python vs other languages (📗 Introduction to Python Statements)

In many languages, blocks are marked with braces and statements end with semicolons:

```javascript
// JavaScript
if (a > b) {
    a = 2;
    b = 4;
}
```

Python uses a **colon `:`** to start a block and **indentation** to define what belongs inside it:

```python
if a > b:
    a = 2
    b = 4
```

Less "noise", more readability. Python's design principle: *code is read far more often than it is written.*

````python
a, b = 10, 5
if a > b:
    print("a is bigger")      # inside the if-block (indented)
    print("still inside")     # inside
print("outside the block")    # not indented → always runs
````

## 2. Indentation rules

- Use **4 spaces** per level (PEP 8 standard). Editors insert 4 spaces when you press Tab.
- All lines in the same block must use the **same** indentation.
- Never mix tabs and spaces.

````python
# ❌ Missing indentation raises IndentationError
if 5 > 2:
print("Five is greater than two!")
````

````python
# ❌ Inconsistent indentation inside one block
if True:
    print("four spaces")
      print("six spaces")
````

> 🆕 **New in Python 3.10+ / 3.13+:** error messages are much friendlier, e.g. *"expected an indented block after 'if' statement on line 2"* and suggestions like *"Did you mean ...?"*.

## 3. Statements and lines

One statement per line is the norm. A semicolon allows several on a line, but **this is discouraged**.

````python
x = 1; y = 2; print(x + y)     # works, but not recommended style
````

### Long lines: implicit continuation inside brackets
Anything inside `()`, `[]` or `{}` can span multiple lines. This is the **preferred** way.

````python
total = (1 + 2 + 3
         + 4 + 5 + 6)
fruits = [
    "apple",
    "banana",
    "cherry",       # a trailing comma is allowed and recommended
]
print(total, fruits)
````

The backslash `\` also continues a line, but it is fragile (a trailing space breaks it), so prefer brackets.

````python
long_sum = 1 + 2 + \
           3 + 4
print(long_sum)
````

## 4. Comments

Comments explain *why* the code does something. Python ignores them.

````python
# This is a single-line comment
price = 100  # an inline comment: two spaces before the #

# Python has no special multi-line comment syntax;
# you simply put # on each line.
print(price)
````

> ⚠️ **Old / misleading (📘 handbook):** the handbook says `'''...'''` is a "multi-line comment". Technically it's a **string literal** that Python creates and throws away. It works, but it is **not** a comment: inside a function or class it becomes a *docstring*. Use `#` for comments, and triple quotes for docstrings.

````python
'''
This "comment" is actually a string that is evaluated and discarded.
'''
print("still runs fine")
````

### Docstrings
A string placed as the **first statement** of a module, function or class is its *docstring*, i.e. documentation that tools and `help()` can read.

````python
def area(radius):
    """Return the area of a circle with the given radius."""
    return 3.14159 * radius ** 2

print(area.__doc__)
help(area)
````

## 5. Keywords (reserved words)

You cannot use these as variable names:

````python
import keyword
print(len(keyword.kwlist), "keywords:")
print(keyword.kwlist)
print("\nSoft keywords (keywords only in certain places):", keyword.softkwlist)
````

**Soft keywords** like `match`, `case`, `type`, `_` (and 🆕 `lazy` in Python 3.15) are only special in specific contexts, so old code using them as names still works.

````python
class = "Python 101"   # ❌ SyntaxError: 'class' is a keyword
````

## 6. PEP 8: the style guide

| Rule | Example |
|---|---|
| 4-space indentation | `    return x` |
| `snake_case` for variables & functions | `total_amount`, `get_user()` |
| `PascalCase` for classes | `BankAccount` |
| `UPPER_CASE` for constants | `MAX_SIZE = 100` |
| Spaces around operators | `x = a + b` not `x=a+b` |
| Max line length 79 (many teams use 88–100) | |
| Two blank lines around top-level functions/classes | |

> 💡 **Tip:** Don't format by hand. Use **Ruff** (`ruff format` and `ruff check`), the modern, very fast linter and formatter, or **Black**. (Covered in Module 07.)
