### 🎯 What you will learn
A **method** is a function that belongs to an object, called with `object.method()`. Strings have ~47 methods; here are the ones you'll use daily, grouped by purpose:

1. Changing case
2. Removing whitespace / prefixes / suffixes
3. Finding, counting & replacing
4. Splitting & joining
5. Checking string properties (`is...` methods)
6. Alignment & padding
7. Character translation

> Remember: strings are immutable, so **every method returns a new string**; the original is unchanged.

````python
text = "hello world"
upper = text.upper()
print(text, "|", upper)     # the original is unchanged
````

## 1. Changing case

````python
s = "hello WORLD from python"
print(s.upper())        # HELLO WORLD FROM PYTHON
print(s.lower())        # hello world from python
print(s.title())        # Hello World From Python
print(s.capitalize())   # Hello world from python
print(s.swapcase())     # HELLO world FROM PYTHON
````

`casefold()` is a more aggressive `lower()` for **case-insensitive comparisons**, including non-English text:

````python
print("Straße".lower(), "Straße".casefold())
print("STRASSE".casefold() == "Straße".casefold())
````

## 2. Removing whitespace and prefixes/suffixes

````python
messy = "   hello world   \n"
print(repr(messy.strip()))     # both sides
print(repr(messy.lstrip()))    # left side
print(repr(messy.rstrip()))    # right side
print("xxhixx".strip("x"))     # strip specific characters
print("www.example.com".strip("w.moc"))   # ❗ strips ANY of these characters, not a word
````

🆕 **New in Python 3.9:** `removeprefix()` / `removesuffix()` remove an exact substring, which is usually what people *meant* by `strip("...")`:

````python
url = "https://python.org"
print(url.removeprefix("https://"))
print("report.csv".removesuffix(".csv"))
print("report.csv".removesuffix(".txt"))    # unchanged if not present
````

## 3. Finding, counting, replacing

````python
s = "Python is fun and Python is powerful"
print(s.find("is"))         # 7: index of first match
print(s.rfind("is"))        # last match
print(s.find("Java"))       # -1 when not found
print(s.index("fun"))       # like find, but raises ValueError when missing
print(s.count("Python"))    # 2
print(s.startswith("Py"), s.endswith("ful"))
print("photo.png".endswith((".png", ".jpg")))  # tuple of options
print(s.replace("Python", "Rust"))
print(s.replace("Python", "Rust", 1))   # only the first occurrence
````

````python
s.index("Java")     # ❌ ValueError: substring not found
````

## 4. Splitting and joining

````python
csv_line = "apple,banana,orange"
fruits = csv_line.split(",")
print(fruits)
print("a  b   c".split())          # no argument → split on any whitespace
print("a,b,c,d".split(",", 2))     # max 2 splits
print("a,b,c,d".rsplit(",", 1))    # split from the right
print("line1\nline2\nline3".splitlines())
print(" - ".join(fruits))          # join is called on the SEPARATOR
print("".join(["P", "y"]))
````

`partition()` splits into exactly 3 parts: before, separator, after:

````python
print("user@example.com".partition("@"))
key, _, value = "name=Asha".partition("=")
print(key, value)
````

## 5. Checking properties (`is...` methods)

These all return `True`/`False`.

````python
tests = ["Python123", "Python", "123", "   ", "hello", "HELLO", "Hello World", "my_var", "½"]
print(f"{'value':<12}{'alpha':>7}{'digit':>7}{'alnum':>7}{'space':>7}{'lower':>7}{'upper':>7}{'title':>7}{'ident':>7}{'numeric':>9}")
for t in tests:
    print(f"{t!r:<12}{t.isalpha()!s:>7}{t.isdigit()!s:>7}{t.isalnum()!s:>7}{t.isspace()!s:>7}"
          f"{t.islower()!s:>7}{t.isupper()!s:>7}{t.istitle()!s:>7}{t.isidentifier()!s:>7}{t.isnumeric()!s:>9}")
````

> 💡 For validating user input like "is this a number?", `isdigit()` fails for `"-5"` and `"3.14"`. The robust way is to try converting with `int()`/`float()` inside `try/except` (Module 07).

## 6. Alignment and padding

````python
word = "Python"
print("[" + word.center(20) + "]")
print("[" + word.center(20, "*") + "]")
print("[" + word.ljust(10, ".") + "]")
print("[" + word.rjust(10, ".") + "]")
print("42".zfill(5))                       # zero padding
print("a\tb\tc".expandtabs(8))
print("Total".ljust(15, ".") + "₹1,250".rjust(10))   # invoice-style
````

## 7. Translating characters

````python
table = str.maketrans("aeiou", "AEIOU")      # map characters
print("hello world".translate(table))
remove_digits = str.maketrans("", "", "0123456789")
print("abc123def456".translate(remove_digits))
````

## 8. Chaining methods

Because each method returns a string, you can chain them:

````python
raw = "   Hello, World!   "
print(raw.strip().lower().replace(",", "").replace("!", "").split())
````

## 9. Useful built-in functions with strings (📘)

````python
s = "Hello, Python!"
print(len(s))
print(sorted("python"))           # list of sorted characters
print(max("python"), min("python"))
print(ord("A"), chr(97))
print(list(reversed("abc")))
````

## Quick reference

| Purpose | Methods |
|---|---|
| Case | `upper lower title capitalize swapcase casefold` |
| Trim | `strip lstrip rstrip removeprefix removesuffix` |
| Search | `find rfind index rindex count startswith endswith in` |
| Modify | `replace translate maketrans expandtabs` |
| Split/Join | `split rsplit splitlines partition rpartition join` |
| Check | `isalpha isdigit isdecimal isnumeric isalnum isspace islower isupper istitle isidentifier isascii isprintable` |
| Align | `center ljust rjust zfill` |
