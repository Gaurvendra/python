## 1. Booleans (📗 Sets and Booleans)

Booleans represent truth values. Note the capital letters: `True`, `False`.

````python
a = True
print(type(a))
print(1 > 2)           # comparisons produce booleans
print(10 == 10.0)
````

`bool` is actually a subclass of `int`: `True == 1` and `False == 0`. Handy for counting:

````python
results = [True, False, True, True]
print(sum(results))        # how many are True → 3
print(True + True)         # 2 (works, but don't write code like this)
````

## 2. Truthiness: what counts as `False`?

In an `if` or `while`, **any** object can be tested. These are **falsy**:

| Falsy values |
|---|
| `False`, `None` |
| `0`, `0.0`, `0j`, `Decimal(0)` |
| empty collections: `""`, `[]`, `()`, `{}`, `set()`, `range(0)` |

**Everything else is truthy**, including `"False"`, `"0"`, `[0]` and `" "`.

````python
for value in [0, 1, -1, "", "0", "False", " ", [], [0], {}, None, 0.0]:
    print(f"{value!r:>8} → {bool(value)}")
````

````python
cart = []
if not cart:                 # Pythonic emptiness check
    print("Your cart is empty")
name = ""
display = name or "Guest"    # 'or' returns the first truthy operand
print(display)
````

## 3. `None`: the absence of a value

`None` is Python's "nothing / no value yet". Functions that don't `return` anything return `None`.

````python
result = None
print(result, type(result))

def say_hi():
    print("hi")

returned = say_hi()
print(returned)
````

Placeholder usage (📗 bootcamp: use `None` as a placeholder for an object we don't want to reassign yet):

````python
b = None
print(b)
````

✅ Always compare with **`is`**, not `==`:

````python
value = None
print(value is None)        # ✅ correct
print(value is not None)
````

> 🆕 **New in Python 3.15 (PEP 661): `sentinel`.** Sometimes `None` is a *valid* value and you need a different "nothing was passed" marker. Python 3.15 adds a built-in `sentinel()` for that (you'll use it in Module 05 · Functions).

````python
MISSING = sentinel("MISSING")
print(MISSING, MISSING is MISSING)
````

## 5. Type conversion (typecasting)

### Explicit conversion: you call the conversion function

````python
# string → number
num_str = "10"
num_int = int(num_str)
print(num_int + 5)                # 15
print(float("3.14"))              # 3.14
print(int("  42  "))              # whitespace is OK
print(int("1_000"))               # underscores OK
print(int("101", 2))              # base 2 → 5

# number → string
print(str(25) + " years")

# float → int (truncates toward zero, does NOT round!)
print(int(3.99), int(-3.99))
print(round(3.99))                # use round() if you want rounding

# to bool
print(bool(0), bool(42), bool(""), bool("hi"))
````

````python
int("3.14")       # ❌ ValueError: you can't go str → int directly with a decimal point
````

````python
print(int(float("3.14")))   # ✅ go through float first
````

````python
int("ten")        # ❌ ValueError
````

### Converting between collections

````python
print(list("abc"))                 # ['a', 'b', 'c']
print(tuple([1, 2, 3]))
print(set([1, 2, 2, 3]))           # removes duplicates
print(list({"a": 1, "b": 2}))      # dict → list of keys
print(dict([("a", 1), ("b", 2)]))  # list of pairs → dict
print(list(range(5)))
print(str([1, 2]))                 # the text "[1, 2]"
````

### Implicit conversion: Python does it for you (only when it's safe)

````python
print(3 + 4.5)        # int is promoted to float → 7.5
print(True + 2)       # bool is treated as int → 3
````

Python **never** silently converts between str and numbers (unlike JavaScript):

````python
"5" + 3    # ❌ TypeError
````

## 6. Safe conversion of user input

````python
def to_int(text, default=0):
    try:
        return int(text)
    except ValueError:
        return default

print(to_int("42"), to_int("abc"), to_int("7", default=-1))
````

## 📝 Summary
- `True`/`False`; `bool` is a subclass of `int`.
- Falsy: `False`, `None`, zeros, empty containers. Everything else is truthy.
- `None` means "no value"; test with `is None`. 🆕 `sentinel()` in 3.15 for custom markers.
- Explicit conversion: `int() float() str() bool() list() tuple() set() dict()`.
- `int()` truncates; use `round()` to round. `int("3.14")` fails; use `int(float(...))`.
