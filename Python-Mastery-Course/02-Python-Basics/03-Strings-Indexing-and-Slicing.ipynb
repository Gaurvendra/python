## 1. Creating strings

A **string** (`str`) is an **ordered, immutable sequence of characters** (Unicode text).

````python
a = 'Hello, Python!'          # single quotes
b = "Hello, World!"           # double quotes, identical behaviour
c = '''This is
a multi-line
string.'''                    # triple quotes keep line breaks
d = "I'm using double quotes so the apostrophe is fine"
e = 'She said "hi"'
print(a, b, c, d, e, sep="\n")
````

````python
' I'm using single quotes, but this will create an error'
````

Fix it by switching quote style or by **escaping** the quote with `\`:

````python
print('I\'m fine now')
````

## 2. Escape sequences

| Sequence | Meaning |
|---|---|
| `\n` | newline |
| `\t` | tab |
| `\\` | a literal backslash |
| `\'` / `\"` | quotes |
| `\u20B9` | Unicode code point (₹) |

````python
print("Hello\nWorld!")
print("Name:\tAsha")
print("C:\\Users\\asha")
print("Price: \u20B9500")
print("\N{SNAKE} \N{GREEK SMALL LETTER PI}")     # Unicode characters by name
````

### Raw strings: `r"..."`
In a raw string, backslashes are **not** escapes. Perfect for Windows paths and regular expressions.

````python
print(r"C:\new\table")       # without r, \n and \t would be interpreted!
print("C:\new\table")
````

## 3. Length

````python
print(len("Hello"))
print(len(""))         # empty string
print(len("I am"))     # spaces count
````

## 4. Indexing

Each character has a position (**index**), starting at **0**. Negative indices count from the end.

```
 P   y   t   h   o   n
 0   1   2   3   4   5     ← positive index
-6  -5  -4  -3  -2  -1     ← negative index
```

````python
text = "Python"
print(text[0])     # P
print(text[1])     # y
print(text[-1])    # n (last character)
print(text[-2])    # o
print(text[len(text) - 1])   # also the last character
````

````python
text[10]     # ❌ IndexError: string index out of range
````

## 5. Slicing `[start:stop:step]`

- `start` is **included**, `stop` is **excluded** ("up to but not including")
- Omitted `start` → beginning; omitted `stop` → end; default `step` = 1
- Slices **never** raise IndexError (they just return fewer characters)

````python
text = "Hello, Python!"
print(text[0:5])     # Hello
print(text[:5])      # Hello
print(text[7:])      # Python!
print(text[-7:-1])   # Python
print(text[:])       # a full copy
print(text[3:100])   # no error even though 100 > len
````

### The step parameter

````python
s = "abcdefghijk"
print(s[::2])        # every 2nd character: acegik
print(s[1::3])       # start at 1, every 3rd: behk
print(s[::-1])       # reverse!  kjihgfedcba
print(s[-1:-5:-1])   # backwards slice: kjih
````

### Practical uses of slicing (📘 handbook)

````python
text = "Welcome to Python!"
print(text[:7])            # first word: Welcome
print(text[-7:])           # last 7 chars: Python!
print(text[3:-3])          # trim 3 from each end
filename = "report_2026.csv"
print(filename[-3:])       # extension: csv
print("racecar" == "racecar"[::-1])   # palindrome check
````

## 6. Strings are immutable

You can't change a character in place; you build a **new** string instead.

````python
name = "Sam"
name[0] = "P"     # ❌ TypeError: 'str' object does not support item assignment
````

````python
name = "Sam"
name = "P" + name[1:]      # create a new string and rebind the name
print(name)
````

## 7. Concatenation, repetition, membership, iteration

````python
first, last = "Ada", "Lovelace"
full = first + " " + last        # concatenation
print(full)
print("-" * 20)                  # repetition
print("Ada" in full, "Bob" not in full)   # membership
for ch in "abc":                 # iteration
    print(ch, end=" ")
````

````python
"Age: " + 30      # ❌ TypeError: can only concatenate str (not "int") to str
````

````python
print("Age: " + str(30))   # convert first...
print(f"Age: {30}")        # ...or (better) use an f-string
````

> 💡 **Performance (📘):** building a big string with `+=` in a loop is slow. Collect pieces in a list and use `"".join(pieces)`.

````python
pieces = []
for i in range(5):
    pieces.append(str(i))
print(",".join(pieces))
````

## 8. Unicode, `ord()`, `chr()`, encoding

Python 3 strings are **Unicode**, so any language and emoji work.

````python
greeting = "नमस्ते 👋 こんにちは"
print(greeting, len(greeting))
print(ord("A"), chr(65))          # character ↔ code point
print(ord("₹"), chr(8377))
````

To save text to a file or send it over the network, it must be **encoded** to bytes. UTF-8 is the universal standard.

````python
data = "café".encode("utf-8")     # str → bytes
print(data, type(data), len(data))
print(data.decode("utf-8"))        # bytes → str
````

> 🆕 **New in Python 3.15 (PEP 686):** **UTF-8 is now the default encoding** for `open()` and other text I/O, regardless of your operating system. Before 3.15, Windows often used a different "locale" encoding, causing bugs. Still: always pass `encoding="utf-8"` explicitly for code that must also run on older Pythons.

````python
s = "Hello\nWorld"
s            # notebook display shows the repr (quotes, escapes)
````

````python
print(s)     # print shows the actual text
````

````python
print(repr(s))   # repr() gives the "developer" representation explicitly
````

## 📝 Summary
- Strings are ordered, immutable sequences of Unicode characters.
- Index from 0 (or -1 from the end); slice with `[start:stop:step]`, stop excluded.
- `[::-1]` reverses a string.
- Use escape sequences or raw strings for special characters.
- Encode to bytes with `.encode()`; UTF-8 is the default in 3.15.
