## 1. Syntax

```python
while condition:
    # code that runs while condition is True
else:
    # optional: runs once when the condition becomes False (not after break)
```

Use `while` when you **don't know in advance** how many iterations are needed.

````python
count = 0
while count < 5:
    print(count)
    count += 1        # ❗ without this line, the loop would never end
````

````python
x = 0
while x < 10:
    print("x is currently:", x, "| x is still less than 10, adding 1 to x")
    x += 1
else:
    print("All Done!")
````

## 2. Infinite loops

If the condition never becomes `False`, the loop runs forever. In Jupyter press ■ **Interrupt** (or `I, I`); in a terminal press `Ctrl + C`.

```python
while True:
    print("This will run forever!")    # ❌ unless there's a break inside
```

An intentional `while True` loop with a `break` is a very common, readable pattern:

````python
import random
random.seed(7)
attempts = 0
while True:
    attempts += 1
    roll = random.randint(1, 6)
    if roll == 6:
        print(f"Rolled a 6 after {attempts} attempts!")
        break
````

## 3. `break`, `continue`, `pass` (📗 + 📘)

| Statement | Effect |
|---|---|
| `break` | Exit the nearest enclosing loop immediately |
| `continue` | Skip the rest of this iteration; go back to the top of the loop |
| `pass` | Do nothing; a placeholder where a statement is required |

````python
x = 0
while x < 10:
    print("x is currently:", x)
    x += 1
    if x == 3:
        print("Breaking because x == 3")
        break
    else:
        print("continuing...")
        continue
    print("this line is never reached")
````

````python
for i in range(5):
    if i == 2:
        continue          # skip 2
    print(i, end=" ")
print()
for i in range(5):
    if i == 3:
        pass              # does nothing, the print still runs
    print(i, end=" ")
print()
````

`pass` is also used for empty functions/classes you'll fill in later:

````python
def todo():
    pass

class EmptyForNow:
    pass

print(todo(), EmptyForNow())
````

> 💡 `...` (Ellipsis) is also commonly used as a placeholder: `def todo(): ...`

## 4. Common `while` patterns

### Countdown

````python
n = 5
while n > 0:
    print(n, end="… ")
    n -= 1
print("Liftoff! 🚀")
````

### Processing until a collection is empty

````python
tasks = ["email", "code review", "meeting"]
while tasks:                       # truthiness: stops when the list is empty
    task = tasks.pop(0)
    print("Doing:", task)
print("All tasks done")
````

### Digit manipulation

````python
n = 9475
digits_sum = 0
while n > 0:
    digits_sum += n % 10     # last digit
    n //= 10                 # remove last digit
print(digits_sum)
````

### Retry with a limit

````python
import random
random.seed(1)
max_retries, attempt = 5, 0
while attempt < max_retries:
    attempt += 1
    success = random.random() > 0.7       # simulate an unreliable network call
    print(f"Attempt {attempt}: {'ok' if success else 'failed'}")
    if success:
        break
else:
    print("Giving up after", max_retries, "attempts")
````

### Menu / game loop (needs `input()`, so run it yourself)

````python
while True:
    choice = input("1) Play  2) Scores  q) Quit: ").strip().lower()
    if choice == "1":
        print("Playing...")
    elif choice == "2":
        print("High score: 42")
    elif choice == "q":
        print("Bye!")
        break
    else:
        print("Invalid choice")
````

### Sentinel loop with the walrus operator (📘)

````python
while (data := input("Enter a value (or 'quit' to exit): ")) != "quit":
    print(f"You entered: {data}")
````

## 5. `for` vs `while`

| Use `for` when... | Use `while` when... |
|---|---|
| iterating over a collection or a known range | the number of iterations is unknown |
| you need each item | you loop until a condition changes (user input, convergence, retries) |

Anything written with `for` can be rewritten with `while`, but `for` is less error-prone (no forgotten increment!).

## 📝 Summary
- `while cond:` repeats while `cond` is truthy; update the state inside the loop!
- `while True:` + `break` is a clean pattern for "loop until".
- `break` exits, `continue` skips, `pass` does nothing.
- `else` on loops runs only if no `break` occurred.
