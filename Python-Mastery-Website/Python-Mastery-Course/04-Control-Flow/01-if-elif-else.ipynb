## 1. Syntax

```python
if condition1:
    # runs if condition1 is True
elif condition2:
    # runs if condition1 is False and condition2 is True
else:
    # runs if all conditions above are False
```

- `elif` and `else` are optional; you can have any number of `elif`s.
- Python checks top-to-bottom and runs **only the first** matching block.

````python
if True:
    print("It was true!")
````

````python
x = False
if x:
    print("x was True!")
else:
    print("I will be printed in any case where x is not true")
````

## 2. Multiple branches (📗 bootcamp + 📘 handbook)

````python
loc = "Bank"
if loc == "Auto Shop":
    print("Welcome to the Auto Shop!")
elif loc == "Bank":
    print("Welcome to the bank!")
else:
    print("Where are you?")
````

````python
age = 18
if age < 18:
    print("You are a minor.")
elif age == 18:
    print("You just became an adult!")
else:
    print("You are an adult.")
````

````python
def grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

for s in [95, 83, 71, 65, 12]:
    print(s, grade(s))
````

❗ **Order matters.** If the `>= 60` test came first, every passing score would get a "D".

## 3. Combining conditions

````python
age, has_id, is_vip = 25, True, False
if age >= 18 and has_id:
    print("Entry allowed")
if is_vip or age > 60:
    print("Priority lane")
if not is_vip:
    print("Regular ticket")
temperature = 23
if 18 <= temperature <= 26:          # chained comparison
    print("Comfortable")
````

## 4. Truthiness in conditions

````python
name = ""
items = ["pen"]
if not name:
    print("Name is empty")
if items:                             # ✅ Pythonic: same as len(items) > 0
    print(f"You have {len(items)} item(s)")
````

> ❌ Avoid `if x == True:` or `if len(items) > 0:`; just write `if x:` / `if items:`.
> ❌ But careful: `if value:` is also False for `0`. If 0 is a valid value, test `if value is not None:`.

## 5. Nested conditions

````python
person = "George"
if person == "Sammy":
    print("Welcome Sammy!")
else:
    if person == "George":
        print("Welcome George!")
    else:
        print("Welcome, what's your name?")
````

Better: flatten with `elif`:

````python
if person == "Sammy":
    print("Welcome Sammy!")
elif person == "George":
    print("Welcome George!")
else:
    print("Welcome, what's your name?")
````

## 6. Guard clauses (early return): a professional habit

Deeply nested `if`s are hard to read. Check for the "bad" cases first and exit early:

````python
# ❌ nested
def withdraw_nested(balance, amount):
    if amount > 0:
        if amount <= balance:
            return balance - amount
        else:
            return "Insufficient funds"
    else:
        return "Invalid amount"

# ✅ guard clauses
def withdraw(balance, amount):
    if amount <= 0:
        return "Invalid amount"
    if amount > balance:
        return "Insufficient funds"
    return balance - amount

print(withdraw(100, 30), withdraw(100, 300), withdraw(100, -5))
````

## 7. One-line conditional expression

````python
age = 16
status = "adult" if age >= 18 else "minor"
print(status)
print("even" if 7 % 2 == 0 else "odd")
````

## 8. `pass`: do nothing (placeholder)

````python
x = 5
if x > 3:
    pass        # TODO: handle this later; an empty block would be a SyntaxError
print("continues")
````

## 9. Indentation (📗)

````python
if True:
    print("inside")
print("outside: always runs")
````

## 📝 Summary
- `if` / `elif` / `else`; only the first true branch runs.
- Combine with `and`, `or`, `not`, and chained comparisons.
- Use truthiness (`if items:`); be careful with 0.
- Prefer `elif` and guard clauses over deep nesting.
- `a if cond else b` for simple choices.
