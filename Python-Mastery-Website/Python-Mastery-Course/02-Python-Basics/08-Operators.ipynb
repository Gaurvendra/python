## 2. Comparison operators (📗 Comparison Operators)

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | equal values | `2 == 2` | `True` |
| `!=` | not equal | `2 != 3` | `True` |
| `>` | greater than | `2 > 3` | `False` |
| `<` | less than | `2 < 3` | `True` |
| `>=` | greater or equal | `2 >= 2` | `True` |
| `<=` | less or equal | `4 <= 1` | `False` |

````python
print(2 == 2, 1 == 0, "Bye" == "bye", "2" == 2)
print(2 != 1, 2 > 1, 2 < 1, 2 >= 2, 4 <= 1)
print(1 == 1.0)        # values compared, not types
````

Strings compare **alphabetically** (by Unicode code point, so uppercase < lowercase):

````python
print("apple" < "banana", "Zebra" < "apple", "10" < "9")   # "10" < "9" because '1' < '9'
````

Lists and tuples compare **element by element**:

````python
print([1, 2, 3] < [1, 2, 4], (1, 9) < (2, 0))
````

````python
print(5 < "5")    # ❌ TypeError: you can't order unrelated types
````

> ⚠️ **Old (Python 2):** `<>` used to mean "not equal". It is a SyntaxError in Python 3; use `!=`.

## 3. Chained comparisons (📗 Chained Comparison Operators)

Python lets you chain comparisons the way you would in maths:

````python
x = 2
print(1 < x < 3)          # same as (1 < x) and (x < 3)
print(1 < x > 3)          # same as (1 < x) and (x > 3) → False
print(0 <= 15 <= 100)     # range check, very readable
a = b = c = 5
print(a == b == c)
````

## 4. Logical operators: `and`, `or`, `not`

| `A` | `B` | `A and B` | `A or B` | `not A` |
|---|---|---|---|---|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

````python
print(True and False, True or False, not True)
age, has_ticket = 20, True
print(age >= 18 and has_ticket)
print(1 == 2 or 2 == 2)
print(not 1 == 1)
````

### Short-circuit evaluation
- `and` stops at the **first falsy** value
- `or` stops at the **first truthy** value

They return **that operand itself**, not necessarily `True`/`False`:

````python
print(0 and 10)          # 0   (stopped at the falsy 0)
print(5 and 10)          # 10  (all truthy → last one)
print("" or "default")   # 'default'
print("Asha" or "default")

user = None
print(user is not None and user.upper())   # safe: second part never runs when user is None
````

## 5. Assignment operators

| Operator | Same as |
|---|---|
| `x = 5` | assign |
| `x += 3` | `x = x + 3` |
| `x -= 3` | `x = x - 3` |
| `x *= 3` | `x = x * 3` |
| `x /= 3` | `x = x / 3` |
| `x //= 3` | `x = x // 3` |
| `x %= 3` | `x = x % 3` |
| `x **= 3` | `x = x ** 3` |
| `x &= ...`, `\|=`, `^=`, `>>=`, `<<=` | bitwise versions |

````python
x = 10
x += 5;  print(x)    # 15
x -= 3;  print(x)    # 12
x *= 2;  print(x)    # 24
x /= 4;  print(x)    # 6.0
x //= 4; print(x)    # 1.0
x **= 3; print(x)    # 1.0
words = ["a"]
words += ["b"]       # works for lists too (extends in place)
print(words)
````

> ❌ Python has **no** `++` / `--` operators. `x++` is a SyntaxError; `++x` silently means `+(+x)`, which does nothing!

## 6. Membership operators: `in`, `not in`

````python
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)
print("grape" not in fruits)
print("py" in "python")                # substring check
print("name" in {"name": "Asha"})      # checks dict KEYS
print(3 in range(10))
print(3 in {1, 2, 3})                  # sets: very fast lookups
````

## 7. Identity operators: `is`, `is not`

- `==` asks: **do they have the same value?**
- `is` asks: **are they the very same object in memory?**

````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a == b, a is b)    # True False: equal values, different objects
print(a == c, a is c)    # True True: same object
print(id(a), id(b), id(c))
````

> ⚠️ **Misleading example in the handbook:**
> ```python
> x = 10
> y = 10
> print(x is y)   # True
> ```
> This prints `True` only because CPython **caches small integers** (-5 to 256 up to Python 3.14; 🆕 **-5 to 1024 in 3.15**) and some strings as an optimisation. The range changed between versions, which proves the point: it's an internal detail. It is an implementation detail, so **never use `is` to compare numbers or strings**:

````python
x = 100_000
y = int("100000")    # created at runtime, outside the small-int cache
print(x == y, x is y)   # True False
````

✅ **Rule:** use `is` only for singletons: `None`, `True`, `False`, and sentinels. Use `==` for everything else.

## 8. Bitwise operators

Work on the binary representation of integers.

| Op | Name | Example (`a=0b1100`=12, `b=0b1010`=10) | Result |
|---|---|---|---|
| `&` | AND | `a & b` | `0b1000` = 8 |
| `\|` | OR | `a \| b` | `0b1110` = 14 |
| `^` | XOR | `a ^ b` | `0b0110` = 6 |
| `~` | NOT | `~a` | `-13` |
| `<<` | shift left | `a << 1` | `24` |
| `>>` | shift right | `a >> 2` | `3` |

````python
a, b = 12, 10
print(a & b, a | b, a ^ b, ~a, a << 1, a >> 2)
print(f"{a:04b} & {b:04b} = {a & b:04b}")
# practical: check if a number is a power of 2
n = 64
print(n > 0 and n & (n - 1) == 0)
````

`|`, `&`, `^`, `-` also work on **sets**, and `|` merges **dicts** (3.9+):

````python
print({1, 2, 3} | {3, 4}, {1, 2, 3} & {2, 3, 4})
print({"a": 1} | {"b": 2})
````

## 9. Conditional (ternary) expression

`value_if_true if condition else value_if_false`

````python
age = 20
status = "adult" if age >= 18 else "minor"
print(status)
n = -7
print("positive" if n > 0 else "zero" if n == 0 else "negative")   # can be chained (use sparingly)
````

## 10. 🆕 The walrus operator `:=` (assignment expression, Python 3.8, 📘 handbook)

`:=` assigns a value **and** returns it, inside a larger expression. Named after the eyes and tusks of a walrus `:=`.

### Use case 1: avoid computing something twice

````python
data = [1, 2, 3, 4, 5, 6, 7]
# without walrus: len() computed twice
if len(data) > 5:
    print(f"List too long ({len(data)} elements)")
# with walrus
if (n := len(data)) > 5:
    print(f"List too long ({n} elements)")
````

### Use case 2: list comprehensions (compute once, filter and keep)

````python
numbers = [1, 2, 3, 4, 5]
results = [x * 2 for x in numbers if x * 2 > 5]             # x * 2 computed twice
results_walrus = [y for x in numbers if (y := x * 2) > 5]    # computed once
print(results, results_walrus)
````

### Use case 3: loops that read until a condition (📘)

````python
import io
fake_file = io.StringIO("line one\nline two\nline three\n")
while (line := fake_file.readline()):
    print(line.strip())
````

Same pattern with `input()` (run it yourself):
```python
while (data := input("Enter a value (or 'quit'): ")) != "quit":
    print(f"You entered: {data}")
```

### Use case 4: regex matching

````python
import re
if (m := re.search(r"\d+", "Order 66 confirmed")):
    print("Found number:", m.group())
````

### Considerations
- ⚠️ **Correction to the handbook:** a variable assigned with `:=` inside a **comprehension** is NOT local to the comprehension; it **leaks into the surrounding scope** (by design):

````python
[y for x in range(3) if (y := x * 10) >= 0]
print(y)    # 20: y is visible after the comprehension
````

- Parentheses are usually required: `(n := len(x)) > 5`. Without them, `n := len(x) > 5` would assign the *boolean*.
- Unparenthesised `:=` at the top level of a statement is a SyntaxError (use `=` there).
- Don't overuse it: readability first.

## 11. Operator precedence (high → low)

| Precedence | Operators |
|---|---|
| 1 | `()` parentheses |
| 2 | `**` |
| 3 | `+x`, `-x`, `~x` (unary) |
| 4 | `*`, `/`, `//`, `%`, `@` |
| 5 | `+`, `-` |
| 6 | `<<`, `>>` |
| 7 | `&` |
| 8 | `^` |
| 9 | `\|` |
| 10 | comparisons, `in`, `not in`, `is`, `is not` |
| 11 | `not` |
| 12 | `and` |
| 13 | `or` |
| 14 | `if ... else` (conditional expression) |
| 15 | `lambda` |
| 16 | `:=` |

````python
print(True or False and False)        # and binds tighter → True or (False and False) → True
print(not 1 == 2)                     # not (1 == 2) → True
print(2 + 3 * 4 ** 2)                 # 2 + 3 * 16 = 50
````

> 💡 When in doubt, **add parentheses**: they make intent obvious to readers.

## 📝 Summary
- Comparisons return booleans and can be **chained** (`0 < x < 10`).
- `and`/`or` short-circuit and return an operand; `not` negates.
- `in` tests membership; `is` tests identity (use only for `None`/`True`/`False`).
- Bitwise operators work on ints, sets (and `|` on dicts).
- `a if cond else b` is the conditional expression.
- `:=` assigns inside expressions; use sparingly.
