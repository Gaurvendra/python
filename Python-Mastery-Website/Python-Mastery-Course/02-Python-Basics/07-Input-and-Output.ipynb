## 1. `print()` in depth

Signature: `print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)`

````python
print("Hello", "World")                    # multiple args → separated by a space
print("Hello", "World", sep=", ", end="!\n")
print("2026", "10", "20", sep="-")
print("Loading", end="")
print("...", end="")
print(" done")
print()                                    # an empty line
print(1, 2.5, True, None, [1, 2])          # any type works, converted with str()
````

### Unpacking a collection with `*`

````python
fruits = ["apple", "banana", "cherry"]
print(fruits)                  # the list itself
print(*fruits)                 # each item as a separate argument
print(*fruits, sep="\n")       # one per line (📗 used in Milestone 2)
````

### Printing to a file or to stderr

````python
import sys
print("This goes to the error stream", file=sys.stderr)
with open("log.txt", "w", encoding="utf-8") as f:
    print("Saved to a file", file=f)
print(open("log.txt", encoding="utf-8").read())
````

### `flush=True` for progress output
Output is usually buffered; `flush=True` forces it to appear immediately (useful for progress bars):

````python
import time
for i in range(5):
    print(f"\rProgress: {i + 1}/5", end="", flush=True)
    time.sleep(0.1)
print()
````

### Pretty-printing big structures

````python
from pprint import pprint
data = {"users": [{"name": "Asha", "roles": ["admin", "editor"], "active": True},
                  {"name": "Ravi", "roles": ["viewer"], "active": False}],
        "count": 2, "settings": {"theme": "dark", "lang": "en"}}
pprint(data, width=60)
````

## 2. `input()`: reading from the keyboard

`input(prompt)` shows the prompt, waits for the user to press Enter, and **always returns a string**.

````python
name = input("Enter your name: ")
age = int(input("Enter your age: "))          # convert the string to int
print(f"Hello {name}, you are {age} years old.")
print(f"Next year you will be {age + 1}.")
````

❌ **Common mistake:** forgetting that `input()` returns a `str`:
```python
age = input("Age: ")     # user types 20
print(age + 1)           # TypeError: can only concatenate str (not "int") to str
```

### Reading several values on one line

````python
a, b = input("Enter two numbers separated by space: ").split()
a, b = int(a), int(b)
print(a + b)

# or, in one go:
nums = list(map(int, input("Enter numbers: ").split()))
print(sum(nums))
````

## 3. Validating user input (📗 Milestone 1 warm-up)

Users type anything. A robust program **loops until the input is valid**.

Below, `input()` is replaced by a *simulated* user (a list of answers) so you can see the logic run. In your own code, simply use `input()`.

````python
simulated_answers = iter(["two", "15", "7"])     # pretend the user typed these
def fake_input(prompt):
    answer = next(simulated_answers)
    print(prompt + answer)
    return answer

def user_choice(ask=fake_input):
    choice = "WRONG"
    acceptable_range = range(0, 11)
    within_range = False

    while not choice.isdigit() or not within_range:
        choice = ask("Please enter a number (0-10): ")
        if not choice.isdigit():
            print("Sorry, that is not a digit!")
        else:
            within_range = int(choice) in acceptable_range
            if not within_range:
                print("Sorry, you are out of acceptable range (0-10)")
    return int(choice)

print("You chose:", user_choice())
````

The same idea, written more simply with `while True` + `try/except` (the modern style you'll learn in Module 07):

````python
def ask_int(prompt, low, high):
    while True:
        try:
            value = int(input(prompt))
        except ValueError:
            print("Please type a whole number.")
            continue
        if low <= value <= high:
            return value
        print(f"Must be between {low} and {high}.")

n = ask_int("Pick 1-10: ", 1, 10)
````

### Yes/No question

````python
def ask_yes_no(prompt):
    while True:
        answer = input(prompt + " (y/n): ").strip().lower()
        if answer in ("y", "yes"):
            return True
        if answer in ("n", "no"):
            return False
        print("Please answer y or n.")

if ask_yes_no("Play again?"):
    print("Let's go!")
````

## 5. Simple display function (📗 Milestone 1 warm-up)

````python
def display(row1, row2, row3):
    print(row1)
    print(row2)
    print(row3)

example_row = [" ", " ", " "]
example_row[1] = "X"
display(example_row, [" "] * 3, [" "] * 3)
````

## 📝 Summary
- `print(*objects, sep, end, file, flush)`; `*list` unpacks items.
- `input()` always returns `str`; convert with `int()`/`float()`.
- Validate input in a loop; never trust users.
- `pprint` for nested data, `getpass` for secrets.
