### 🎯 What you will learn
String formatting = inserting values into text. Python has grown **four** ways over the years:

| Style | Introduced | Status |
|---|---|---|
| `"%s" % value` | Python 1 | ⚠️ **Old**, still seen in legacy code & `logging` |
| `"{}".format(value)` | Python 2.6 / 3.0 | OK, useful for templates stored as strings |
| `f"{value}"` (f-strings) | Python 3.6 | ✅ **Recommended** for everyday code |
| `t"{value}"` (t-strings) | 🆕 Python 3.14 | Advanced: safe templating (HTML, SQL, logs) |

Plus the **format specification mini-language** (`:>10.2f` etc.), which all of them share.

## 1. ⚠️ Old style: `%` placeholders (📗 bootcamp)

You'll meet this in older code, so you should be able to read it.

````python
name, age = "Alice", 30
print("My name is %s and I am %d years old." % (name, age))
print("I'm going to inject %s here." % "something")
print("Pi is about %.2f" % 3.14159)          # 2 decimal places
print("Padded: %10.2f|" % 13.144)            # width 10, precision 2
print("%s vs %r" % ("hi", "hi"))             # %r uses repr() → adds quotes
print("He said %s" % "tab\there")
print("He said %r" % "tab\there")
````

Conversion types: `%s` (str), `%r` (repr), `%d` (integer, truncates), `%f` (float), `%x` (hex), `%%` (a literal `%`).

````python
print("%d%%" % 3.75)        # truncates 3.75 to 3, then prints a %
````

> ⚠️ **Why it's considered old:** clumsy with many values, tuple-vs-single-value pitfalls, and limited features. **Exception:** the `logging` module still deliberately uses `%`-style (`logger.info("user %s", name)`) so formatting only happens if the message is actually logged.

## 2. The `.format()` method

````python
print("My name is {} and I am {} years old.".format(name, age))
print("{1} is learning {0}".format("Python", "Alice"))        # by index
print("{0} {0} {1}".format("ha", "!"))                         # reuse
print("{name} is {age} years old".format(name="Bob", age=25))  # by keyword
person = {"name": "Eve", "city": "Delhi"}
print("{name} lives in {city}".format(**person))               # from a dict
````

`.format()` is still the right tool when the **template is stored separately** from the data (e.g. loaded from a config file or translation file):

````python
TEMPLATE = "Dear {customer}, your order #{order_id} has shipped."
print(TEMPLATE.format(customer="Ravi", order_id=1042))
````

## 3. ✅ f-strings (formatted string literals)

Put `f` before the quotes and write any **expression** inside `{}`.

````python
name, age = "Alice", 30
print(f"My name is {name} and I am {age} years old.")
x, y = 10, 5
print(f"The sum of {x} and {y} is {x + y}")
print(f"Upper: {name.upper()}, length: {len(name)}")
print(f"Is adult? {'yes' if age >= 18 else 'no'}")
items = ["pen", "book"]
print(f"First item: {items[0]}, count: {len(items)}")
````

### 🆕 Self-documenting expressions with `=` (Python 3.8+), great for debugging

````python
total = 42
print(f"{total=}")
print(f"{x + y = }")
print(f"{name.upper()=}")
````

### 🆕 Python 3.12 (PEP 701): f-strings got more flexible
- You can reuse the **same quote type** inside the braces
- Expressions can span **multiple lines** and contain **comments**
- Backslashes are allowed inside the expressions

````python
songs = ["Take me back to Eden", "Alkaline", "Ascensionism"]
print(f"Playlist: {", ".join(songs)}")     # same " quotes inside, only valid from 3.12
print(f"{'\n'.join(songs)}")               # backslash inside {}, only valid from 3.12
print(f"Total: {
    sum([10, 20, 30])  # a comment inside an f-string!
}")
````

## 4. The format specification mini-language

Syntax inside braces: `{value:[fill][align][sign][width][,][.precision][type]}`

| Spec | Meaning | Example | Result |
|---|---|---|---|
| `.2f` | 2 decimal places | `f"{3.14159:.2f}"` | `3.14` |
| `10` | width 10 | `f"{'hi':10}"` | `hi        ` |
| `<` `>` `^` | left / right / center | `f"{'hi':^10}"` | `    hi    ` |
| `*^10` | fill character | `f"{'hi':*^10}"` | `****hi****` |
| `,` or `_` | thousands separator | `f"{1234567:,}"` | `1,234,567` |
| `+` | always show sign | `f"{5:+}"` | `+5` |
| `%` | percentage | `f"{0.256:.1%}"` | `25.6%` |
| `e` | scientific | `f"{12345.678:.2e}"` | `1.23e+04` |
| `b o x X` | binary / octal / hex | `f"{255:x}"` | `ff` |
| `08.2f` | zero-pad | `f"{3.5:08.2f}"` | `00003.50` |

````python
pi = 3.14159265
print(f"Pi rounded: {pi:.2f}")
text = "Python"
print(f"[{text:>10}]")   # right align
print(f"[{text:<10}]")   # left align
print(f"[{text:^10}]")   # center
print(f"[{text:-^20}]")  # center with '-' fill
print(f"{1234567.891:,.2f}")
print(f"{0.0725:.1%}")
print(f"{42:+} {-42:+}")
print(f"{255:b} {255:o} {255:x} {255:#x}")
print(f"{7:03d}")
````

**Nested format specs:** width and precision can themselves be variables (📗 *"result: {value:{width}.{precision}}"*):

````python
value, width, precision = 13.14159, 10, 4
print(f"result: {value:{width}.{precision}}")
````

**Conversions** `!r`, `!s`, `!a`:

````python
word = "café"
print(f"{word!s} | {word!r} | {word!a}")
````

**Dates** have their own format codes:

````python
from datetime import datetime
now = datetime(2026, 10, 20, 14, 30)
print(f"{now:%d %B %Y, %I:%M %p}")
print(f"{now:%Y-%m-%d}")
````

### Example: printing a neat table

````python
products = [("Laptop", 1, 54999.0), ("Mouse", 2, 499.5), ("USB-C Cable", 3, 199.0)]
print(f"{'Item':<15}{'Qty':>5}{'Price':>12}")
print("-" * 32)
for name, qty, price in products:
    print(f"{name:<15}{qty:>5}{price:>12,.2f}")
print("-" * 32)
print(f"{'TOTAL':<15}{'':>5}{sum(q * p for _, q, p in products):>12,.2f}")
````

## 5. 🆕 Template strings: t-strings (Python 3.14, PEP 750)

An f-string immediately produces a finished `str`. A **t-string** (`t"..."`) produces a **`Template` object** that keeps the static text and the interpolated values **separate**. A library can then process the values safely, for example escaping HTML to prevent injection attacks, or turning values into SQL parameters.

> This is an advanced feature, mostly for **library authors**. As a learner, just know it exists and why.

````python
from string.templatelib import Template, Interpolation

name = "World"
template = t"Hello {name}!"
print(type(template))
print(template.strings)          # the static parts
print(template.values)           # the interpolated values
for part in template:            # iterate over parts in order
    print(repr(part))
````

### Example: a tiny safe-HTML renderer

````python
import html

def render_html(template: Template) -> str:
    parts = []
    for item in template:
        if isinstance(item, Interpolation):
            parts.append(html.escape(str(item.value)))   # escape user data
        else:
            parts.append(item)                            # trust the literal text
    return "".join(parts)

user_input = "<script>alert('hacked')</script>"
print(f"<p>{user_input}</p>")                # ❌ f-string: dangerous
print(render_html(t"<p>{user_input}</p>"))   # ✅ t-string: escaped
````

## 6. `string.Template`, the simple `$`-substitution (different from t-strings!)

Useful for **user-supplied templates** because it can't run arbitrary expressions.

````python
from string import Template as SimpleTemplate
s = SimpleTemplate("Hi $name, you owe $$${amount}")
print(s.substitute(name="Kiran", amount=50))
````

## 📝 Summary
- ⚠️ `%`-formatting is old; know how to read it (still used by `logging`).
- `.format()` is good for templates stored separately from the data.
- ✅ Use **f-strings** for everyday formatting; `f"{x=}"` for debugging.
- 3.12 made f-strings more flexible (quotes, multi-line, comments).
- 🆕 3.14 **t-strings** return `Template` objects for safe processing.
- The format mini-language (`:>10,.2f`) works in f-strings, `.format()` and `format()`.
