## 1. What is a variable?

A variable is a **name that refers to a value (an object)** stored in memory. You create one with the assignment operator `=`.

```
   name           object in memory
  ┌──────┐       ┌───────────────┐
  │ age  │ ────► │  int: 25      │
  └──────┘       └───────────────┘
```

````python
name = "Alice"      # str
age = 25            # int
height = 5.6        # float
is_student = True   # bool

print(name, age, height, is_student)
````

> 💡 Think of a variable as a **label / sticky note** on an object, not as a box that contains the value. Two names can label the same object (important later for lists!).

````python
a = [1, 2, 3]
b = a            # b is a second label on the SAME list
b.append(4)
print(a)         # a sees the change too!
print(a is b)    # True: same object
````

## 2. Naming rules (must follow)

1. Can contain letters, digits and underscores `_`
2. Must **start with a letter or underscore**, never a digit
3. **Case-sensitive**: `age`, `Age` and `AGE` are three different variables
4. Can't be a keyword (`if`, `for`, `class`, ...)
5. No spaces or symbols like `- ! @ # $ %`

````python
user_name = "ok"
_private = "ok"
name2 = "ok"
Age = 30
age = 25
print(Age, age)   # different variables
````

````python
2name = "invalid"   # ❌ starts with a digit
````

````python
first-name = "invalid"   # ❌ '-' is the minus operator
````

### Best practices (PEP 8)
- Use **descriptive** names: `total_price`, not `tp` or `x1`
- `snake_case` for variables/functions, `UPPER_CASE` for constants, `PascalCase` for classes
- Avoid single letters except for short loop counters (`i`, `j`) or math
- ❌ **Never shadow built-ins** such as `list`, `str`, `dict`, `sum`, `max`, `input`, `type`:

````python
list = [1, 2, 3]          # ❌ shadows the built-in list()
numbers = list("abc")     # TypeError: 'list' object is not callable
````

````python
del list                  # remove our variable; the built-in is visible again
print(list("abc"))
````

## 3. Dynamic typing (📗 bootcamp)

Python is **dynamically typed**: you don't declare types, and a name can be re-bound to a value of a different type.

````python
my_dogs = 2
print(type(my_dogs))
my_dogs = ["Sammy", "Frankie"]   # same name, now a list
print(type(my_dogs))
````

In a **statically typed** language (C, Java), this would be an error:
```c
int my_dog = 1;
my_dog = "Sammy";   // compile error
```

#### ✅ Pros of dynamic typing
- Very easy to work with, less boilerplate
- Faster development and prototyping

#### ⚠️ Cons of dynamic typing
- May cause unexpected bugs at run-time
- You need to be aware of `type()`

> 💡 **Modern Python gives you the best of both:** optional **type hints** checked by tools like *mypy* or *pyright*, while still running dynamically. (Full lesson: Module 05 · Type Hints.)

````python
count: int = 10            # a type hint: documentation + checkable by tools
price: float = 9.99
tags: list[str] = ["new", "sale"]
print(count, price, tags)
````

## 4. Reassignment and updating

````python
a = 5
a = 10          # reassign
a = a + 10      # use the old value to compute the new one
a += 10         # shorthand (augmented assignment), same as a = a + 10
print(a)
````

## 5. Multiple assignment, swapping, unpacking

````python
x, y, z = 1, 2, 3            # multiple assignment
print(x, y, z)

p = q = r = 0                # same value to several names
print(p, q, r)

x, y = y, x                  # swap without a temporary variable
print("after swap:", x, y)
````

````python
first, *middle, last = [10, 20, 30, 40, 50]    # extended unpacking
print(first, middle, last)
````

````python
a, b = [1, 2, 3]     # ❌ ValueError: too many values to unpack
````

## 6. Inspecting variables: `type()`, `isinstance()`, `id()`

````python
value = 3.14
print(type(value))                     # <class 'float'>
print(type(value) == float)
print(isinstance(value, float))        # preferred for type checks
print(isinstance(value, (int, float))) # check against several types
print(id(value))                       # unique identity (memory address in CPython)
````

## 7. Constants

Python has no `const` keyword. By **convention**, names in `UPPER_CASE` are treated as constants: please don't change them.

````python
PI = 3.14159
MAX_USERS = 100
from typing import Final
API_URL: Final = "https://api.example.com"   # type checkers will flag reassignment
print(PI, MAX_USERS, API_URL)
````

## 8. Deleting a name

````python
temp = 42
del temp
print(temp)    # NameError: name 'temp' is not defined
````

## 9. Example: putting it together (📗 bootcamp simple exercise)

````python
my_income = 100
tax_rate = 0.1
my_taxes = my_income * tax_rate
print(f"Income: {my_income}, tax rate: {tax_rate:.0%}, taxes: {my_taxes}")
````

## 📝 Summary
- A variable is a name bound to an object; `=` binds, it doesn't copy.
- Follow the naming rules; use `snake_case`, descriptive names; don't shadow built-ins.
- Python is dynamically typed; optional type hints add safety.
- Use tuple unpacking for multiple assignment and swapping.
- `isinstance()` is the preferred way to check types.
