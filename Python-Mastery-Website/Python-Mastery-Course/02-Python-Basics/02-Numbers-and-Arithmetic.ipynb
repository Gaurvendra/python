## 1. Number types

| Type | Example | Notes |
|---|---|---|
| `int` | `7`, `-3`, `1_000_000` | **Unlimited size** in Python |
| `float` | `3.14`, `-0.5`, `2e-3` | 64-bit double precision |
| `complex` | `2 + 3j` | Real + imaginary parts |
| `bool` | `True`, `False` | Subclass of `int` (`True == 1`) |
| `decimal.Decimal` | `Decimal("0.1")` | Exact decimal math (money!) |
| `fractions.Fraction` | `Fraction(1, 3)` | Exact rational numbers |

````python
print(type(7), type(3.14), type(2 + 3j), type(True))
big = 2 ** 200                  # ints never overflow
print(big)
million = 1_000_000             # underscores improve readability
print(million)
````

## 2. Arithmetic operators

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | addition | `7 + 2` | `9` |
| `-` | subtraction | `7 - 2` | `5` |
| `*` | multiplication | `7 * 2` | `14` |
| `/` | true division (always float) | `7 / 2` | `3.5` |
| `//` | floor division | `7 // 2` | `3` |
| `%` | modulo (remainder) | `7 % 2` | `1` |
| `**` | exponent | `7 ** 2` | `49` |

````python
print(2 + 1)
print(2 - 1)
print(2 * 2)
print(3 / 2)       # true division → 1.5
print(4 / 2)       # still a float → 2.0
print(7 // 4)      # floor division: "how many whole times"
print(7 % 4)       # modulo: the remainder
print(2 ** 3)      # power
print(4 ** 0.5)    # square root via power
````

> ⚠️ **Old (Python 2):** in Python 2, `3 / 2` gave `1`. In Python 3, `/` always returns a float. Use `//` for integer division.

### Floor division with negative numbers
`//` rounds **down** (towards −∞), not towards zero:

````python
print(7 // 2, -7 // 2)      # 3  -4
print(7 % 3, -7 % 3)        # 1  2  (result has the sign of the divisor)
print(divmod(17, 5))        # (quotient, remainder) in one call
````

### Modulo is incredibly useful

````python
n = 17
print("even" if n % 2 == 0 else "odd")    # even/odd test
print(125 % 10)                           # last digit
seconds = 3725
print(f"{seconds // 3600}h {seconds % 3600 // 60}m {seconds % 60}s")
````

## 3. Order of operations (precedence)

Highest to lowest: `()` → `**` → unary `-` → `* / // %` → `+ -`. (Think *PEMDAS*.)

````python
print(2 + 10 * 10 + 3)      # 105
print((2 + 10) * (10 + 3))  # 156
print(-2 ** 2)              # -4  (** binds tighter than unary minus!)
print((-2) ** 2)            # 4
print(2 ** 3 ** 2)          # 512  (** is right-associative: 2 ** 9)
````

## 4. Floating-point precision

Computers store floats in binary, and many decimals (like 0.1) can't be stored exactly.

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

**How to compare floats safely:**

````python
import math
print(math.isclose(0.1 + 0.2, 0.3))
print(round(0.1 + 0.2, 10) == 0.3)
````

**For money, use `Decimal`** (always create it from a *string*):

````python
from decimal import Decimal, ROUND_HALF_UP
print(Decimal("0.1") + Decimal("0.2"))
print(Decimal(0.1))   # ❌ from a float: inherits the float's error!
price = Decimal("19.995")
print(price.quantize(Decimal("0.01"), rounding=ROUND_HALF_UP))
````

**For exact fractions, use `Fraction`:**

````python
from fractions import Fraction
third = Fraction(1, 3)
print(third + third + third)      # exactly 1
print(Fraction("0.75"))           # 3/4
````

### Special float values

````python
print(float("inf"), -float("inf"), float("nan"))
print(math.inf > 10**100)
print(math.nan == math.nan)       # NaN is never equal to anything, even itself
print(math.isnan(float("nan")))
````

## 5. Rounding, `abs`, `pow` (📗 Advanced Numbers)

````python
print(round(3.14159, 2))   # 3.14
print(round(395, -2))      # 400  (negative digits round to tens/hundreds)
print(round(3))            # 3
print(abs(-3.14))          # 3.14
print(pow(3, 4))           # 81
print(pow(3, 4, 5))        # (3**4) % 5 = 1, fast for big numbers
print(pow(3, -1, 7))       # modular inverse (3.8+) → 5, since 3*5 % 7 == 1
````

> ❌ **Common surprise: banker's rounding.** `round()` rounds **half to even**, which reduces bias in statistics:

````python
print(round(0.5), round(1.5), round(2.5), round(3.5))   # 0 2 2 4
print(round(2.675, 2))   # 2.67, not 2.68! (2.675 is really 2.67499999... in binary)
````

Other ways to round:

````python
print(math.floor(3.7), math.ceil(3.2), math.trunc(-3.7), int(-3.7))
````

## 6. Other number bases & notation (📗 Advanced Numbers)

````python
print(bin(10), oct(10), hex(255))     # to strings
print(0b1010, 0o12, 0xFF)             # literals
print(int("ff", 16), int("1010", 2))  # parse strings in a base
print(f"{255:b} {255:o} {255:x} {255:08b}")   # via f-strings
print(1.5e3, 2e-4)                    # scientific notation
print((255).bit_length())             # number of bits needed
````

## 7. Complex numbers (📗 complex())

````python
z = complex(2, 3)        # or 2 + 3j
print(z, z.real, z.imag, abs(z))
print(z * (1 - 1j))
print(complex("10+2j"))
````

## 8. Converting between number types

````python
print(int(3.99))       # 3  (truncates toward zero)
print(int("42"))       # 42
print(float(7))        # 7.0
print(float("3.5"))    # 3.5
print(int(True))       # 1
````

> 🆕 **New in Python 3.11:** converting a huge string to `int` (more than 4300 digits) raises `ValueError` to prevent denial-of-service attacks. Raise the limit with `sys.set_int_max_str_digits()` if you really need to.

> 🆕 **New in Python 3.14:** `float.from_number()` and `complex.from_number()` convert *numbers only* (they reject strings, unlike `float()`).

````python
print(float.from_number(5))
try:
    float.from_number("5")
except TypeError as e:
    print("TypeError:", e)
````

## 9. Handy built-ins & `math` quick tour

````python
nums = [4, 9, 1, 7]
print(sum(nums), min(nums), max(nums), sum(nums) / len(nums))
print(math.sqrt(16), math.pi, math.e)
print(math.factorial(5), math.gcd(12, 18), math.lcm(4, 6))
print(math.prod([1, 2, 3, 4]))    # product of an iterable
print(math.isqrt(17))             # integer square root
print(math.dist((0, 0), (3, 4)))  # distance between points
````

## 📝 Summary
- `int` (unlimited), `float` (approximate), `complex`; `Decimal` & `Fraction` for exactness.
- `/` is true division, `//` floors, `%` gives remainder, `**` powers.
- Use parentheses to make precedence explicit; remember `-2**2 == -4`.
- Never compare floats with `==`; use `math.isclose()`.
- `round()` uses banker's rounding.
