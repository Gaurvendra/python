### 🟠 Deprecated functions & APIs (warn today)

| Deprecated | Since | Use instead |
|---|---|---|
| `datetime.utcnow()`, `utcfromtimestamp()` | 3.12 | `datetime.now(UTC)`, `fromtimestamp(ts, UTC)` |
| `typing.List`, `Dict`, `Tuple`, `Set`, `Optional`, `Union` | 3.9/3.10 (soft) | `list`, `dict`, `tuple`, `set`, `X \| None`, `X \| Y` |
| `typing.TypeVar` + `Generic` boilerplate | (style) | `def f[T](...)`, `class C[T]:` (3.12) |
| `from __future__ import annotations` | (no longer needed from 3.14) | deferred annotations are the default |
| `profile` module | 3.15 (removal 3.17) | `profiling.tracing` / `cProfile` |
| `re.match()` | soft-deprecated 3.15 | `re.prefixmatch()` (same behaviour) |
| `NamedTuple("P", x=int)` | 🔴 removed 3.15 | class syntax or list of tuples |
| `asyncio.get_event_loop()` with no running loop | 3.12+ (error in 3.14) | `asyncio.run()`, `asyncio.get_running_loop()` |
| `return`/`break`/`continue` in `finally` | 3.14 `SyntaxWarning` | restructure the code |
| `tarfile.extractall()` without `filter` | default changed in 3.14 | `filter="data"` |
| `unittest` `assertEquals`, `failUnless`... aliases | 🔴 removed 3.12 | `assertEqual`, `assertTrue` |
| `configparser.SafeConfigParser`, `readfp` | 🔴 removed 3.12 | `ConfigParser`, `read_file` |

### 🟡 Old idioms ↔ modern equivalents

| Old | Modern |
|---|---|
| `"Hi %s" % name`, `"Hi {}".format(name)` | `f"Hi {name}"` |
| `os.path.join(a, b)`, `open(os.path...)` | `Path(a) / b`, `Path.read_text()` |
| `try: d[k] except KeyError` everywhere | `d.get(k, default)`, `defaultdict` |
| `class P: __init__ + __repr__ + __eq__` | `@dataclass` |
| constants as strings `"PENDING"` | `enum.Enum` / `StrEnum` |
| `if/elif` on data shape | `match`/`case` |
| `pdb.set_trace()` | `breakpoint()` |
| `time.time()` for durations | `time.perf_counter()` |
| `pytz` | `zoneinfo` |
| `requests` in async code | `httpx.AsyncClient` / `aiohttp` |
| `setup.py install`, `easy_install` | `pip install .`, `uv` |
| `PyPDF2` | `pypdf` |
| `Image.ANTIALIAS` (Pillow) | `Image.Resampling.LANCZOS` |
| `flake8 + isort + black` | `ruff check` + `ruff format` |
| `pip freeze > requirements.txt` only | `pyproject.toml` + `uv lock` |
| `git checkout` for everything | `git switch` / `git restore` |
| `master` default branch | `main` |

### 🔴 Python 2 leftovers (instant red flags in tutorials)
`print "x"` · `raw_input()` · `xrange()` · `dict.has_key()` · `dict.iteritems()` · `unicode()` · `<>` · `except Exception, e:` (Python 2 syntax; note that Python 3.14's `except A, B:` means something different: two exception types) · `3 / 2 == 1` · backticks for `repr` · `reduce` as a built-in.

## How to find deprecations in your own code

```python
code = '''
import datetime, typing
def f(x: typing.List[int]) -> typing.Optional[int]:
    print("value: %s" % x)
    return datetime.datetime.utcnow()
'''
with open("legacy.py", "w", encoding="utf-8") as fh:
    fh.write(code)

import subprocess, sys
print(subprocess.run([sys.executable, "-m", "ruff", "check", "legacy.py", "--select", "UP,DTZ",
                      "--no-cache", "--output-format", "concise"], capture_output=True, text=True).stdout)
```

- `python -W error::DeprecationWarning your_script.py` turns deprecation warnings into errors
- `ruff check --select UP --fix` (pyupgrade rules) rewrites many old idioms automatically
- Read the *Porting to Python 3.x* section of each *What's New* page before upgrading
