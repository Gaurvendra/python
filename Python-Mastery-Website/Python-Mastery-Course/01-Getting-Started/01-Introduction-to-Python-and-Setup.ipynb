## 1. What is programming?

**Programming** means giving a computer a precise list of instructions. The computer does exactly what you say, not what you *meant*, so instructions must be unambiguous.

A **programming language** is a human-readable way to write those instructions. Python's **interpreter** reads your code and executes it line by line.

```
Your code (hello.py)  ──►  Python interpreter  ──►  Bytecode  ──►  Python Virtual Machine runs it  ──►  Output
```

## 2. Why Python?

| Strength | What it means for you |
|---|---|
| **Readable** | Code looks almost like English; indentation is part of the syntax |
| **High-level** | No manual memory management, no compiling step |
| **Interpreted & interactive** | Try ideas instantly in a REPL or notebook |
| **"Batteries included"** | Huge standard library: files, JSON, dates, HTTP, testing, ... |
| **Huge ecosystem** | 600k+ packages on PyPI |
| **Multi-paradigm** | Procedural, object-oriented and functional styles |

**Where Python is used:**
- Web development: Django, Flask, FastAPI
- Data science & AI/ML: pandas, NumPy, scikit-learn, PyTorch, TensorFlow
- Automation & scripting: files, emails, spreadsheets, web scraping
- DevOps & cloud tooling, testing, security
- Games (Pygame), desktop apps (Tkinter, PySide/Qt)

## 3. Installing Python

1. Go to **https://www.python.org/downloads/** and download the latest stable version for your OS.
   - At the time this course was updated (Oct 2026): **Python 3.15.0** is the newest release (9 Oct 2026); **3.14** is the previous stable line.
2. **Windows:** tick ✅ **"Add python.exe to PATH"** in the installer.
3. **macOS:** use the python.org installer, or `brew install python`.
4. **Linux:** usually pre-installed; otherwise `sudo apt install python3`.

### Check the installation
Open a terminal (Command Prompt / PowerShell / Terminal) and type:
```bash
python --version      # Windows
python3 --version     # macOS / Linux
```
You should see something like `Python 3.15.0`.

> 💡 **Tip – the modern way: `uv`.** Many developers now use [`uv`](https://docs.astral.sh/uv/) to install and switch Python versions:
> ```bash
> uv python install 3.15
> uv run --python 3.15 hello.py
> ```
> You'll learn more about environments in **Module 08**.

> ⚠️ **Deprecated / Old:** **Python 2** reached end-of-life on 1 Jan 2020. If a tutorial uses `print "hello"` (no parentheses), `raw_input()`, or `xrange()`, it is Python 2: don't learn from it.

You can always check the version **from inside Python**:

````python
import sys
print(sys.version)          # full version string
print(sys.version_info)     # structured: major, minor, micro, ...
print(sys.version_info >= (3, 10))   # handy for feature checks
````

## 5. Four ways to run Python

### 5.1 The interactive REPL (Read-Eval-Print Loop)
Type `python` (or `python3`) in a terminal:
```text
>>> 2 + 3
5
>>> exit()
```
> 🆕 **New in Python 3.13:** the REPL now has **colours, multi-line editing, history browsing**, and you can simply type `exit` or `help` without parentheses.

### 5.2 Running a script file
Create `hello.py`:
```python
print("Hello, World!")
```
Run it from the terminal (or press ▶ in VS Code):
```bash
python hello.py
```

### 5.3 A Jupyter notebook (like this one)
Each **cell** can be code or text (Markdown). Run a cell with **Shift + Enter**.

### 5.4 One-liners with `-c` and modules with `-m`
```bash
python -c "print(2 ** 10)"
python -m http.server 8000     # run a standard-library module as a program
```

## 6. Your first program

`print()` is a **built-in function** that displays output.

````python
print("Hello, World!")
````

````python
print("Python executes code line by line,")
print("from top to bottom.")
````

In a notebook, the **last expression** of a cell is displayed automatically even without `print()`. In a `.py` script it is **not**: always use `print()` in scripts.

````python
"This is shown because it is the last line of the cell"
````

````python
"This line is NOT shown"
"Only this last one is"
````
