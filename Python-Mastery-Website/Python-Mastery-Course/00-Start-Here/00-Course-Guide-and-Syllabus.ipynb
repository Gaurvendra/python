# 🐍 Python Mastery Course: Guide & Syllabus

**Welcome!** This course takes you from your first `print()` to **advanced, professional Python** (internals, packaging, concurrency, validation, deployment). It's built as the **union** of two sources, then extended and updated for **Python 3.15** (released 9 October 2026):

- 📗 **Complete Python 3 Bootcamp** (Pierian Data): ~110 Jupyter notebooks
- 📘 **Python Handbook** (PDF, 115 pages): fundamentals → OOP → advanced → files/OS → libraries → Git

Every topic from both sources is included (see `01-Coverage-Map`), outdated content is **marked and fixed** (Module 17 · 02), and everything an advanced Python developer needs that was missing has been **added**: internals & metaprogramming, advanced typing, the full packaging toolchain (`pyproject.toml`, **uv**, **Poetry**, wheels, **twine**), deep concurrency & asyncio, advanced testing, professional workflow, SQLAlchemy, and a complete **Pydantic** module.

## How each lesson is organised
1. 🎯 **What you will learn**
2. Explanations in plain language, with tables and diagrams
3. **Many runnable examples**, already executed with Python 3.15 so you can read real outputs even before running them
4. Callouts: 🆕 new in Python 3.x · ⚠️ deprecated/old · ❌ common mistake · 💡 tip · 📗/📘 source
5. 📝 **Summary** and 🏋️ **Practice** (solutions hidden under *Show solutions*)
6. Larger exercises in each module's **`exercises/`** folder and projects in **Module 18**

## Setup (once)
```bash
# Install Python 3.15 (python.org) or: uv python install 3.15
cd Python-Mastery-Course
uv venv && source .venv/bin/activate        # or: python3 -m venv .venv && source .venv/bin/activate
uv pip install -r requirements.txt          # or: python -m pip install -r requirements.txt
jupyter lab                                  # or open the folder in VS Code and pick the .venv kernel
```
> 💡 A few cells use Python 3.15-only features (`frozendict`, `sentinel`, `lazy import`, `[*x for x in ...]`) and are labelled. Everything else runs on Python 3.12+. Notebooks that run tools (uv, Poetry, Git, pip) do so in **temporary sandbox folders**, so they never change your real environments.
