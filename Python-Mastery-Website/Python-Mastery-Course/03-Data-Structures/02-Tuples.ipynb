## 1. Constructing tuples

A tuple is an **ordered, immutable** sequence. It uses parentheses, but it's really the **comma** that makes a tuple.

````python
t = (1, 2, 3)
mixed = ("one", 2, 3.0)
no_parens = 4, 5, 6            # also a tuple (packing)
empty = ()
single = (5,)                  # ✅ a 1-element tuple needs a trailing comma
not_a_tuple = (5)              # ❌ just the int 5 in parentheses
print(type(t), type(no_parens), type(single), type(not_a_tuple))
print(len(t), tuple("abc"), tuple([1, 2]))
````

## 2. Indexing & slicing

````python
my_tuple = (10, 20, 30, 40)
print(my_tuple[1], my_tuple[-1], my_tuple[1:3], my_tuple[::-1])
````

## 3. Tuple methods: only two!

| Method | Description | Example | Output |
|---|---|---|---|
| `count(x)` | how many times `x` appears | `(1, 2, 2, 3).count(2)` | `2` |
| `index(x)` | index of the first `x` | `(10, 20, 30).index(20)` | `1` |

````python
my_tuple = (1, 2, 2, 3, 4)
print(my_tuple.count(2))
print(my_tuple.index(3))
````

## 4. Immutability

Once created, a tuple's items can't be reassigned, added or removed.

````python
t = ("one", 2, 3)
t[0] = "change"       # ❌ TypeError
````

````python
t.append("nope")      # ❌ AttributeError: tuples have no append
````

To "change" a tuple, build a new one:

````python
t = ("one", 2, 3)
t = ("ONE",) + t[1:]
print(t)
````

❗ **Subtle point:** a tuple is immutable, but if it *contains* a mutable object (like a list), that object can still change:

````python
t = (1, [2, 3])
t[1].append(4)
print(t)
````

## 5. Packing and unpacking

````python
point = (3, 4)            # packing
x, y = point              # unpacking
print(x, y)

a, b, c = (10, 20, 30)
print(a, b, c)

first, *rest = (1, 2, 3, 4)
print(first, rest)        # note: rest becomes a LIST

# swapping is tuple packing + unpacking
a, b = b, a
print(a, b)
````

### Returning multiple values from a function

````python
def min_max(numbers):
    return min(numbers), max(numbers)    # returns a tuple

low, high = min_max([4, 9, 1, 7])
print(low, high)
````

### Looping over pairs

````python
work_hours = [("Abby", 100), ("Billy", 400), ("Cassie", 800)]
for name, hours in work_hours:
    print(f"{name} worked {hours} hours")
````

## 6. When to use tuples (📗 + 📘)

Use a **tuple** when:
- The data is a **fixed record** whose positions mean something: `(lat, lon)`, `(r, g, b)`, `(name, age)`
- You need a **dictionary key** or a **set** member: tuples are **hashable** (if their contents are)
- You want to **guarantee** that the data won't be changed accidentally ("data integrity")

Use a **list** when the collection will **grow, shrink or change**.

````python
distances = {("Delhi", "Mumbai"): 1400, ("Pune", "Mumbai"): 150}
print(distances[("Pune", "Mumbai")])
````

````python
bad = {["Delhi", "Mumbai"]: 1400}   # ❌ lists are unhashable → can't be keys
````

> ⚠️ **Overstated in the handbook:** *"tuples are faster than lists"*. Tuples are slightly smaller in memory and a bit faster to create, but for everyday code the difference is negligible. **Choose based on meaning (fixed vs changing), not speed.**

````python
import sys
print(sys.getsizeof((1, 2, 3)), sys.getsizeof([1, 2, 3]))
````

## 7. Preview: named tuples

When positions get confusing (`person[2]`?), give the fields names:

````python
from collections import namedtuple
Point = namedtuple("Point", ["x", "y"])
p = Point(3, 4)
print(p, p.x, p.y, p[0])
````

## 📝 Summary
- Tuples: ordered, immutable, can hold duplicates and mixed types.
- `(5,)` is a tuple, `(5)` is not.
- Only two methods: `count`, `index`.
- Great for fixed records, multiple return values, dict keys.
