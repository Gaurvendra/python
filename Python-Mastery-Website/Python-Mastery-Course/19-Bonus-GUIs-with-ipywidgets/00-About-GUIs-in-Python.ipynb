## Other GUI options in Python (2026)

| Goal | Library |
|---|---|
| Widgets inside notebooks | **ipywidgets** (this module) |
| Quick data apps / dashboards in the browser | **Streamlit**, **Gradio**, **Panel**, **NiceGUI** |
| Desktop apps (built in) | **tkinter** (ships with Python) |
| Professional desktop apps | **PySide6 / PyQt6** (Qt) |
| Beautiful terminal (TUI) apps | **Textual**, **Rich** |
| Web apps | **FastAPI**, **Flask**, **Django** + HTML/JS |

A minimal `tkinter` window (run it as a `.py` file):
```python
import tkinter as tk

root = tk.Tk()
root.title("Hello")
tk.Label(root, text="Hello from tkinter!").pack(padx=20, pady=10)
tk.Button(root, text="Quit", command=root.destroy).pack(pady=10)
root.mainloop()
```
> ⚠️ Python 3.15's release notes mention that on macOS 27.0, IDLE and tkinter apps may hang when opening dialogs (CPython issue #158053).
