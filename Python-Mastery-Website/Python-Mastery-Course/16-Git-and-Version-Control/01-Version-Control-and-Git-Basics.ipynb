## 1. What is version control?

**Version control** is a system that records changes to files over time so you can recall specific versions later, like a **time machine for your code**.

**Why use it?**
- Tracks every change: who, what, when and **why**
- Restore older versions when something breaks
- Lets many people work on the same project simultaneously
- Experiment safely on branches

**Without version control:**
```
index.html
index_final.html
index_final_final.html
index_final_v2_really_final.html
```
Messy, error-prone and impossible to collaborate on.

## 2. Centralised vs distributed

| Centralised (CVCS), e.g. SVN | Distributed (DVCS), e.g. **Git** |
|---|---|
| one central server holds the history | **every** developer has the full history |
| server down → nobody can commit | commit offline; sync later |
| single point of failure | every clone is a backup |

## 3. Git vs GitHub

- **Git**: the **tool** that tracks your code history, installed on your computer. Created by **Linus Torvalds** (creator of Linux) in **2005**.
- **GitHub**: a **hosting service** for Git repositories (like Google Drive for Git projects), plus pull requests, issues, Actions (CI/CD)... Alternatives: **GitLab**, **Bitbucket**, Codeberg.

## 5. Configuring Git (once per computer)

```bash
git config --global user.name "Your Name"
git config --global user.email "you@example.com"
git config --global init.defaultBranch main          # new repos start on "main"
git config --global core.editor "code --wait"        # optional: VS Code as editor
git config --list                                    # check your settings
```
Your name and email are stored in every commit. Use the same email as your GitHub account.

> ⚠️ **Old default:** older Git versions name the first branch **`master`**. GitHub and most projects now use **`main`**; set `init.defaultBranch main` as above.

## 6. The three areas of Git

```
 Working Directory  ──git add──►  Staging Area  ──git commit──►  Repository (.git)
 (your files)                     (next snapshot)                (permanent history)
```
1. **Working directory**: where you edit files
2. **Staging area** (*index*): where you prepare exactly what goes into the next commit
3. **Repository**: where commits (snapshots) are stored permanently

File states: **untracked** → **staged** → **committed** → (edit) **modified** → staged → ...

## 7. First Git commands

### Creating a repository: `git init`

````python
%%bash
cd "$PLAYGROUND"
mkdir my-first-repo && cd my-first-repo
git init -b main
ls -a                      # the hidden .git folder is the "brain": it stores the entire history
````

The other way to start is to **clone** an existing repository (downloads the whole project *with its history*):
```bash
git clone https://github.com/user/repo.git
```

### Adding a file and checking the status

````python
%%bash
cd "$PLAYGROUND/my-first-repo"
echo "Hello Git!" > hello.txt
git status
````

**Untracked** means Git sees the file but isn't tracking it yet.

### Staging changes: `git add`

````python
%%bash
cd "$PLAYGROUND/my-first-repo"
git add hello.txt          # stage one file   (git add . stages everything in the folder)
git status
````

### Committing: `git commit`

A **commit** is a snapshot of the staged changes, with a message.

````python
%%bash
cd "$PLAYGROUND/my-first-repo"
git commit -m "Add hello.txt with a greeting"
````

✅ **Good commit messages** describe *why* the change was made, in the imperative mood: `Fix login crash when email is empty`, not `changes` or `stuff`.

### Viewing history: `git log`

````python
%%bash
cd "$PLAYGROUND/my-first-repo"
git log
echo "-----"
git log --oneline
````

## 8. Making further changes

````python
%%bash
cd "$PLAYGROUND/my-first-repo"
echo "This is my first change" >> hello.txt
git status --short           # M = modified (compact view)
git add hello.txt
git commit -q -m "Update hello.txt with a new line"
echo "This is another line" >> hello.txt
git commit -q -am "Add another line"     # -a stages all TRACKED modified files, then commits
git log --oneline
````

## 9. Quick command recap (📘)

| Command | Purpose |
|---|---|
| `git init` | start a new repository |
| `git clone <url>` | copy a repository |
| `git status` | check file states |
| `git add <file>` / `git add .` | stage changes |
| `git commit -m "msg"` | save staged changes |
| `git commit -am "msg"` | stage tracked files + commit |
| `git log --oneline` | view history in short form |
