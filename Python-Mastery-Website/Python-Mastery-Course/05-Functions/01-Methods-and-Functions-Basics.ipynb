## 1. Methods (📗)

**Methods** are functions that *belong to an object*. Call them with `object.method(arguments)`.

````python
lst = [1, 2, 3, 4, 5]
lst.append(6)
print(lst.count(2), lst)
````

How do I find out which methods exist?
- In Jupyter: type `lst.` and press **Tab**
- Press **Shift+Tab** inside parentheses to see the signature
- Use `dir()` and `help()`:

````python
print([m for m in dir(lst) if not m.startswith("_")])
help(lst.count)
````

## 2. What is a function and why use one?

A **function** is a named, reusable block of code that performs one task.

**Why?**
- **Reusability**: write once, call many times (DRY: *Don't Repeat Yourself*)
- **Modularity**: split a big problem into small, testable pieces
- **Abstraction**: callers don't need to know *how* it works, just *what* it does
- **Readability**: `calculate_tax(income)` explains itself

## 3. Defining and calling

```python
def function_name(parameter1, parameter2):
    """Docstring: what the function does."""
    # body
    return result
```

````python
def say_hello():
    print("hello")

say_hello()       # ✅ call it with parentheses
say_hello         # ❌ without (), you just reference the function object
````

````python
def greet(name):
    print(f"Hello {name}")

greet("Jose")
greet("Asha")
````

**Parameters vs arguments:** `name` in the definition is a *parameter*; `"Jose"` in the call is an *argument*.

````python
greet()     # ❌ TypeError: missing 1 required positional argument: 'name'
````

## 4. `return`: sending a value back

````python
def add_num(num1, num2):
    return num1 + num2

result = add_num(4, 5)
print(result)
print(add_num("one", "two"))     # works on strings too (duck typing!)
````

### ❓ "What is the difference between `return` and `print`?" (📗 very common question)

- `print()` **displays** a value on the screen. The function's result is still `None`.
- `return` **hands a value back** to the caller so it can be stored and used later.

````python
def print_result(a, b):
    print(a + b)

def return_result(a, b):
    return a + b

x = print_result(10, 20)    # prints 30...
y = return_result(10, 20)   # prints nothing...
print("x =", x)             # ...but x is None
print("y =", y)             # y holds 30
print(return_result(1, 2) * 10)   # a returned value can be used in expressions
````

`return` also **ends the function immediately**:

````python
def check(n):
    if n < 0:
        return "negative"      # leaves here
    return "non-negative"
    print("never runs")

print(check(-1), check(5))
````

A function without `return` (or with a bare `return`) returns **`None`**.

## 5. Adding logic inside functions (📗)

````python
def even_check(number):
    return number % 2 == 0      # the comparison already IS a boolean

print(even_check(20), even_check(21))
````

````python
# Return True if ANY number in a list is even
def check_even_list(num_list):
    for number in num_list:
        if number % 2 == 0:
            return True
    return False                # ❗ AFTER the loop, not in an else inside it

print(check_even_list([1, 3, 5]), check_even_list([1, 2, 5]))
````

❌ **Common bug:**
```python
for number in num_list:
    if number % 2 == 0:
        return True
    else:
        return False      # returns after checking only the FIRST number!
```

````python
# Return ALL even numbers in a list
def all_even(num_list):
    even_numbers = []
    for number in num_list:
        if number % 2 == 0:
            even_numbers.append(number)
    return even_numbers

print(all_even([1, 2, 3, 4, 5, 6]))
print(any(n % 2 == 0 for n in [1, 3, 4]))     # the built-in way
````

## 6. Returning tuples for unpacking (📗)

````python
work_hours = [("Abby", 100), ("Billy", 400), ("Cassie", 800)]

def employee_check(work_hours):
    current_max = 0
    employee_of_month = ""
    for employee, hours in work_hours:
        if hours > current_max:
            current_max = hours
            employee_of_month = employee
    return employee_of_month, current_max      # a tuple

name, hours = employee_check(work_hours)
print(name, hours)
````

## 7. Functions calling functions (📗 three-cup Monte)

Breaking a program into small functions that call each other:

````python
from random import shuffle, seed

def shuffle_list(mylist):
    shuffle(mylist)          # shuffles in place
    return mylist

def player_guess(simulated=None):
    guess = ""
    while guess not in ["0", "1", "2"]:
        guess = simulated if simulated else input("Pick a number: 0, 1, or 2: ")
    return int(guess)

def check_guess(mylist, guess):
    if mylist[guess] == "O":
        print("Correct! The ball was there:", mylist)
    else:
        print("Wrong guess!", mylist)

seed(3)
cups = [" ", "O", " "]
mixed_up = shuffle_list(cups)
guess = player_guess(simulated="1")      # remove simulated= to play for real
check_guess(mixed_up, guess)
````

## 8. Docstrings (📘 handbook)

A docstring is the first string inside a function. Tools, IDEs and `help()` show it.

````python
def add(a, b):
    """
    Return the sum of two numbers.

    Parameters:
        a (int | float): The first number.
        b (int | float): The second number.

    Returns:
        int | float: The sum of a and b.

    Example:
        >>> add(2, 3)
        5
    """
    return a + b

print(add.__doc__)
````

````python
help(add)
````

> 💡 Popular docstring styles: **Google style** (above), **NumPy style**, **reStructuredText**. Pick one and be consistent. With **type hints** (lesson 06) you can drop the types from the docstring.

## 9. Functions are objects ("first-class citizens")

You can assign them to variables, store them in lists/dicts, and pass them to other functions. This is the foundation of decorators (Module 09).

````python
def shout(text):
    return text.upper() + "!"

yell = shout                 # no parentheses: we're referencing, not calling
print(yell("hi"))
print(shout.__name__)

def apply(func, value):
    return func(value)

print(apply(shout, "python"))
print(apply(len, "python"))
actions = {"shout": shout, "length": len}
print(actions["length"]("hello"))
````

## 10. Naming & design guidelines
- Use verbs: `calculate_total`, `send_email`, `is_valid` (booleans: `is_`, `has_`)
- One function = one job. If you need "and" to describe it, split it.
- Prefer **returning** values over printing inside functions (easier to test and reuse).
- Keep functions short (roughly < 20–30 lines).

## 📝 Summary
- Methods belong to objects; functions are defined with `def`.
- Parameters in the definition, arguments in the call.
- `return` sends a value back (and exits); `print` only displays. No return → `None`.
- Return multiple values as a tuple and unpack them.
- Write docstrings; functions are first-class objects.
