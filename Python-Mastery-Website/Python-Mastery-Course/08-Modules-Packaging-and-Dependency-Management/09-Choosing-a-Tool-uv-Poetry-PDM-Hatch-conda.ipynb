## 1. The jobs to be done

| Job | Tools |
|---|---|
| Install Python versions | python.org, OS package managers, **uv**, pyenv, conda, Poetry (🧪) |
| Create environments | `venv`, virtualenv, **uv**, Poetry, PDM, Hatch, conda |
| Install packages | **pip**, **uv**, conda |
| Declare dependencies | **`pyproject.toml`** (all modern tools), `requirements.in` |
| Lock dependencies | **uv.lock**, **poetry.lock**, pdm.lock, `pylock.toml` (PEP 751), pip-tools `requirements.txt`, conda-lock |
| Run CLI tools in isolation | **pipx**, **uv tool / uvx** |
| Build distributions | `python -m build`, **uv build**, Poetry, Hatch, PDM, flit |
| Publish | **twine**, **uv publish**, Poetry, Hatch, PDM, flit |
| Task running / matrices | Hatch envs, nox, tox, `poe the poet`, `just`, Makefiles |

## 2. Comparison

| | pip + venv | **uv** | **Poetry** | PDM | Hatch | conda / pixi |
|---|---|---|---|---|---|---|
| Speed | ok | ⚡ fastest | ok | good (can use uv) | good (can use uv) | ok / fast (pixi) |
| Python version management | ❌ | ✅ | 🧪 | ✅ | ✅ | ✅ |
| Lock file | ❌ (pip-tools) | ✅ universal | ✅ | ✅ | ⚠️ limited | ✅ (conda-lock / pixi.lock) |
| Standard `[project]` metadata | ✅ | ✅ | ✅ (Poetry 2) | ✅ | ✅ | pixi: ✅ |
| PEP 735 dependency groups | pip ≥ 25.1 | ✅ | ✅ | ✅ | via envs | – |
| Scripts with inline deps (PEP 723) | ❌ | ✅ | ❌ | ✅ | ✅ | ❌ |
| Global tools (pipx-like) | pipx | ✅ | ❌ | ❌ | ❌ | pixi global |
| Workspaces / monorepo | ❌ | ✅ | plugins | ✅ | ❌ | ✅ |
| Build + publish | build + twine | ✅ | ✅ | ✅ | ✅ | – |
| Non-Python deps (CUDA, GDAL, R) | ❌ | ❌ | ❌ | ❌ | ❌ | ✅ |
| Learning curve | lowest | low | medium | medium | medium | medium |

## 3. Recommendations (2026)

| Situation | Recommendation |
|---|---|
| Learning Python, small scripts | `python -m venv` + `pip`, or simply **uv** |
| New application or library | ✅ **uv** (`uv init`, `uv add`, `uv.lock`, `uv build`, `uv publish`) |
| Existing Poetry codebase | keep **Poetry** (upgrade to 2.x + `[project]` format), or migrate with `uvx migrate-to-uv` |
| Single-file automation scripts | **uv** with PEP 723 metadata (`uv run script.py`) |
| Data science with heavy native deps / GPUs | **conda (Miniforge)** or **pixi**; uv is also excellent for pure-PyPI stacks |
| Testing across many Python versions | **nox** or **tox** (both can use uv), or Hatch environments |
| Personal CLI tools | `uv tool install` / `pipx` |
| Compiled extensions | maturin (Rust) / scikit-build-core (C/C++) / setuptools + **cibuildwheel** |

## 4. Migration paths

| From → To | How |
|---|---|
| `requirements.txt` → uv | `uv init --bare && uv add -r requirements.txt` |
| `requirements.txt` → Poetry | `poetry init -n` then `poetry add $(cat requirements.txt)` |
| Poetry / Pipenv / PDM → uv | `uvx migrate-to-uv` |
| `setup.py` → `pyproject.toml` | move metadata into `[project]`; keep `setup.py` only for C extensions |
| Poetry 1.x → Poetry 2.x | move `[tool.poetry]` metadata into `[project]`; dev deps into groups; `poetry lock` |
| conda → uv | export the pure-Python dependencies (`conda env export --from-history`), then `uv add` them |

## 5. A modern project, end to end (uv)

```bash
uv init --lib textstats && cd textstats
uv add regex
uv add --dev pytest ruff mypy
uv run pytest && uv run ruff check . && uv run mypy src
uv version --bump minor
uv build && uv publish          # (trusted publishing in CI)
```

## 📝 Summary
- `pyproject.toml` is the common foundation; tools differ in speed and scope.
- **uv** is the recommended default for new projects; **Poetry** remains a solid, mature choice; **conda/pixi** when you need non-Python binaries.
- Use pipx/`uv tool` for CLI apps, nox/tox for test matrices, cibuildwheel for compiled wheels.
